# Aviso 2.0 playground: Destination Earth data notifications

This notebook queries the **Aviso 2.0 playground server** with the **aviso client**
(`pyaviso` 2.x, the Rust-based rewrite from <https://github.com/ecmwf/aviso-client>)
and works through the Destination Earth (DestinE) data streams it publishes.

| Setting | Value |
| --- | --- |
| Server | `https://aviso-playground.ecmwf.int` |
| DestinE event types | `de-scm-data`, `de-scm-data-points` |
| Auth | Bearer token from <https://api.ecmwf.int/v1/key/> (needed for replay and watch, not for the schema listing) |
| Client | `pip install pyaviso` (2.1.0 at the time of writing; provides `import pyaviso` and the `aviso` CLI) |

What we cover:

1. Connect and authenticate
2. Discover the schema of the DestinE streams
3. Replay the stored history (which data landed where on the data bridge)
4. Narrow by experiment, date, and publish time
5. Inspect the CloudEvent envelope
6. Spatial filter on the `de-scm-data-points` stream
7. Listen live (time-bounded so the notebook does not block forever)
8. Triggers: write an NDJSON log per notification
9. Resume across restarts with a state file
10. Publish a test notification (opt-in)
11. Error handling
12. CLI equivalents

> **Aviso 1.x on the LUMI data bridge.** The DestinE Extremes-DT `data` event on
> `aviso.lumi.apps.dte.destination-earth.eu` is still served by the *old* etcd-based
> Aviso 1.x server and needs the old client (`pip install 'pyaviso==1.0.2'`). The 2.x
> client in this notebook cannot talk to it. See `../test/DESTINE_GEN2_LUMI_SETUP.md`
> and `../DE/aviso-examples/` for that endpoint. The playground is where the DestinE
> streams on the **new** server live today.

## 1. Connect and authenticate

The token is read from `AVISO_TOKEN` if set, otherwise from the CLI configuration file
`~/.config/aviso/config.yaml` (`auth.bearer_token`). It is never printed.

If you do not have either yet:

```bash
export AVISO_TOKEN="<token from https://api.ecmwf.int/v1/key/>"
```

````python
# Uncomment to install into the active kernel:
# %pip install -q pyaviso pyyaml pandas

import json
import os
import pathlib
import threading
import time
from datetime import datetime, timedelta, timezone

import pandas as pd
import pyaviso

print("pyaviso", pyaviso.__version__)
````

````python
BASE_URL = "https://aviso-playground.ecmwf.int"
EVENT = "de-scm-data"               # DestinE stream on the playground
POINTS_EVENT = "de-scm-data-points"  # companion stream with a spatial identifier
CLI_CONFIG = pathlib.Path.home() / ".config" / "aviso" / "config.yaml"


def resolve_token() -> str:
    # AVISO_TOKEN wins; otherwise fall back to the aviso CLI config file.
    token = os.environ.get("AVISO_TOKEN")
    if token:
        return token
    if CLI_CONFIG.exists():
        import yaml

        cfg = yaml.safe_load(CLI_CONFIG.read_text(encoding="utf-8")) or {}
        token = (cfg.get("auth") or {}).get("bearer_token")
        if token:
            return token
    raise RuntimeError(
        "No aviso token found. Set AVISO_TOKEN or add auth.bearer_token to "
        f"{CLI_CONFIG} (get a token from https://api.ecmwf.int/v1/key/)."
    )


token = resolve_token()
client = pyaviso.AvisoClient(base_url=BASE_URL, auth=pyaviso.Bearer(token))
print("connected to", client.base_url, f"| token loaded ({len(token)} chars)")
````

### A small compatibility helper

`pyaviso` 2.1.0 names the start cursor `start_from=`; the development branch of
`aviso-client` renames it to `from_=`. The helper below tries both so the rest of the
notebook reads the same on either version. Everything else on `client.listen(...)` is
identical: `filter=`, `mode=`, `triggers=`.

````python
def open_stream(client, event_type, *, filter, start=None, mode=None, triggers=None):
    kwargs = {"filter": filter}
    if mode is not None:
        kwargs["mode"] = mode
    if triggers is not None:
        kwargs["triggers"] = triggers
    if start is None:
        return client.listen(event_type, **kwargs)
    try:
        return client.listen(event_type, start_from=start, **kwargs)
    except TypeError:
        return client.listen(event_type, from_=start, **kwargs)


def replay(client, event_type, *, filter, start=1, triggers=None):
    # Drain the stored history from `start` and return it as a list.
    with open_stream(
        client, event_type, filter=filter, start=start, mode="replay_only", triggers=triggers
    ) as stream:
        return list(stream)
````

## 2. Discover the schema

The client never defines schemas, it discovers them. `client.schema()` lists every event
type on the server; `client.schema_for(...)` returns the identifier fields of one stream
and which of them a **watch or replay must commit to** (`required: true`).

````python
catalog = client.schema()
print("status:", catalog.status, "| event types on the playground:", catalog.total_schemas)
for name in sorted(catalog.event_types):
    tag = "  <- DestinE" if name.startswith("de-") else ""
    print(f"  {name}{tag}")
````

````python
for name in (EVENT, POINTS_EVENT):
    schema = client.schema_for(name).schema
    print(f"=== {name} ===")
    print(json.dumps(schema, indent=2))
    print()
````

Reading the `de-scm-data` schema:

| Identifier | Type | Required on watch/replay | Notes |
| --- | --- | --- | --- |
| `experiment` | `StringHandler` (max 10 chars) | yes | an experiment id such as `jab0`; `"*"` acts as a wildcard |
| `date` | `DateHandler` (`YYYYMMDD`) | no | omit to match every date |
| `time` | `TimeHandler` (`HHMM`) | no | omit to match every time |

The payload is required for publishers. On this stream it is a JSON object with an
`output_path`, the location where the experiment output landed (a file system path or an
`ec:/...` archive path). That is the "which data is where" information.

String fields do **not** accept constraint objects (`{"gte": ...}` and friends), only a
scalar value or the `"*"` wildcard. Section 11 shows the server's error for that case.

## 3. Replay the stored history

`mode="replay_only"` streams what the server still holds, starting at a cursor, and stops
after the server's `replay_completed` control event. Sequence `1` means "from the
beginning of whatever is retained". Older sequences may have aged out of retention, so the
first sequence you see is not necessarily 1.

````python
history = replay(client, EVENT, filter={"experiment": "*"}, start=1)
print(f"{len(history)} notifications retained on {EVENT}")
for n in history:
    print(f"seq={n.sequence:<4} {n.identifier}  ->  {n.payload}")
````

````python
def to_frame(notifications):
    rows = []
    for n in notifications:
        ce = n.cloudevent or {}
        rows.append(
            {
                "sequence": n.sequence,
                "experiment": n.identifier.get("experiment"),
                "date": n.identifier.get("date"),
                "time": n.identifier.get("time"),
                "output_path": (n.payload or {}).get("output_path") if isinstance(n.payload, dict) else n.payload,
                "published_utc": ce.get("time"),
            }
        )
    frame = pd.DataFrame(rows)
    if not frame.empty:
        frame["published_utc"] = pd.to_datetime(frame["published_utc"], utc=True, format="ISO8601")
    return frame


df = to_frame(history)
df
````

````python
if not df.empty:
    display(df.groupby("experiment").agg(notifications=("sequence", "size"),
                                         first_date=("date", "min"),
                                         last_date=("date", "max"),
                                         last_published=("published_utc", "max")))
````

## 4. Narrow the replay

Every identifier key you add to `filter=` narrows the match. The start cursor can be a
sequence id **or** an ISO 8601 UTC timestamp. A timestamp cursor refers to the **publish
time** of the notification on the server, not to the `date` field of the identifier
(the forecast or experiment date). Filter on `date` for the latter.

````python
experiments = sorted({n.identifier.get("experiment") for n in history})
print("experiments seen:", experiments)

# Pick the most recent experiment that has real output (skip obvious test placeholders).
real = df[~df["output_path"].fillna("").str.startswith("/path/to")] if not df.empty else df
EXPERIMENT = real.sort_values("published_utc")["experiment"].iloc[-1] if not real.empty else (experiments[-1] if experiments else "*")
print("using experiment:", EXPERIMENT)

by_experiment = replay(client, EVENT, filter={"experiment": EXPERIMENT}, start=1)
to_frame(by_experiment)
````

````python
# Same experiment, one specific date (identifier date, YYYYMMDD)
if by_experiment:
    some_date = by_experiment[-1].identifier["date"]
    one_day = replay(client, EVENT, filter={"experiment": EXPERIMENT, "date": some_date}, start=1)
    print(f"{EXPERIMENT} on {some_date}: {[n.sequence for n in one_day]}")
````

````python
# Publish-time cursor: everything published in the last 30 days, any experiment.
since = datetime.now(timezone.utc) - timedelta(days=30)
cursor = since.strftime("%Y-%m-%dT%H:%M:%SZ")
recent = replay(client, EVENT, filter={"experiment": "*"}, start=cursor)
print(f"published since {cursor}: {len(recent)} notifications")
to_frame(recent)
````

## 5. The CloudEvent envelope

Each notification is delivered as a CloudEvent. `Notification` exposes the narrowed view
(`event_type`, `sequence`, `identifier`, `payload`) and keeps the raw envelope on
`.cloudevent`. The envelope carries the publish timestamp (`time`), the stream schema URL
(`dataschema`) and the event id `<event_type>@<sequence>`.

````python
if history:
    last = history[-1]
    print(json.dumps(last.cloudevent, indent=2))
    print()
    print("as_dict():", json.dumps(last.as_dict())[:200], "…")
````

## 6. Spatial filter on `de-scm-data-points`

Publishers on `de-scm-data-points` attach a `point_cloud` (a list of `[lat, lon]` points).
Subscribers must **not** send `point_cloud`; they send a closed `polygon` instead and
receive the notifications whose point cloud intersects it. The polygon is a list of
`[lat, lon]` pairs with the first point repeated at the end.

````python
europe = [[30, -30], [30, 45], [75, 45], [75, -30], [30, -30]]   # lat, lon; closed ring
points = replay(client, POINTS_EVENT, filter={"experiment": "*", "polygon": europe}, start=1)
print(f"{len(points)} notifications on {POINTS_EVENT} intersecting the Europe box")
for n in points[:5]:
    pc = n.identifier.get("point_cloud")
    print(f"seq={n.sequence} experiment={n.identifier.get('experiment')} date={n.identifier.get('date')} "
          f"points={len(pc) if isinstance(pc, list) else pc} payload={n.payload}")
if not points:
    print("(the stream is currently empty or nothing intersects the box; the call shape is what matters here)")
````

## 7. Listen live, time-bounded

`client.listen(event, filter=...)` without a mode is a live watch: the iterator blocks until
the next matching notification arrives and reconnects on its own. A notebook cell cannot
block forever, so the helper below closes the iterator from a timer thread after a
deadline or after `max_items` notifications, whichever comes first.

Pass `start=` to replay the history first and then continue live on the same stream. That
is the standard way to catch up after a disconnect without missing anything.

````python
def listen_for(client, event_type, *, filter, seconds=20, max_items=None, start=None, triggers=None):
    stream = open_stream(client, event_type, filter=filter, start=start, triggers=triggers)
    timer = threading.Timer(seconds, stream.close)
    timer.start()
    received = []
    t0 = time.monotonic()
    try:
        for n in stream:
            received.append(n)
            print(f"[{time.monotonic() - t0:5.1f}s] seq={n.sequence} {n.identifier} -> {n.payload}")
            if max_items and len(received) >= max_items:
                break
    finally:
        timer.cancel()
        stream.close()
    print(f"watch closed after {time.monotonic() - t0:.1f}s with {len(received)} notification(s)")
    return received


live = listen_for(client, EVENT, filter={"experiment": "*"}, seconds=15)
````

````python
# Replay the last few from history, then stay live for a short while.
if history:
    cursor = max(history[-3].sequence - 1, 0)   # sequences after this one
    caught_up = listen_for(client, EVENT, filter={"experiment": "*"}, start=cursor, seconds=10)
````

## 8. Triggers: act on every notification

Triggers are side effects the client runs for each notification **before** yielding it:
`echo` (stdout), `log` (append NDJSON to a file), `command` (shell), `webhook`, `teams`,
`post`. They compose with replay and watch alike. Below, a replay writes the whole
retained history of the stream into an NDJSON file in `./output/`.

````python
OUT_DIR = pathlib.Path("output")
OUT_DIR.mkdir(exist_ok=True)
log_path = OUT_DIR / f"{EVENT}.ndjson"
log_path.unlink(missing_ok=True)

logged = replay(
    client, EVENT, filter={"experiment": "*"}, start=1,
    triggers=[pyaviso.Trigger.log(log_path)],
)
print(f"{len(logged)} notifications, {log_path} now has {sum(1 for _ in log_path.open())} line(s)")
print(log_path.read_text(encoding="utf-8")[:600])
````

````python
# A command trigger: fields are available as {{ notification.<path> }} templates and as AVISO_* env vars.
marker = OUT_DIR / "seen.txt"
marker.unlink(missing_ok=True)
cmd = pyaviso.Trigger.command(
    'echo "{{ notification.identifier.experiment }} {{ notification.identifier.date }} '
    '-> {{ notification.payload.output_path }}" >> ' + str(marker.resolve()),
    timeout=10.0,
)
_ = replay(client, EVENT, filter={"experiment": "*"}, start=1, triggers=[cmd])
print(marker.read_text(encoding="utf-8"))
````

## 9. Resume across restarts

Attach a `JsonFileStore` and the client commits the last delivered sequence per
(server, event type, filter) key. The next run with the same filter resumes after that
cursor instead of starting at the live edge. Run the cell twice: the second run delivers
only what arrived since the first one.

`flush_cursor_on_exit=True` makes sure the very last notification is committed when the
iterator is closed (by default a cursor is committed only when the *next* notification is
sent).

````python
state_path = OUT_DIR / "state.json"
resuming = pyaviso.AvisoClient(
    base_url=BASE_URL,
    auth=pyaviso.Bearer(token),
    state_store=pyaviso.JsonFileStore(state_path),
    flush_cursor_on_exit=True,
)

# First run: no cursor yet, so `start=1` replays the retained history and commits the cursor.
# Later runs: the committed cursor takes over.
resumed = listen_for(resuming, EVENT, filter={"experiment": "*"}, start=1, seconds=8)
print(json.dumps(json.loads(state_path.read_text(encoding="utf-8")), indent=2))
````

## 10. Publish a test notification (opt-in)

The playground is a shared sandbox, so this cell is disabled by default. Set
`PUBLISH_TEST = True` to publish one notification. A publish must supply **every**
identifier field of the schema (`experiment`, `date`, `time`) plus the required payload,
regardless of the `required` flag, which only governs watch and replay filters.

Run the live listener from section 7 in a second kernel first if you want to see the
publish arrive in real time.

````python
PUBLISH_TEST = False

if PUBLISH_TEST:
    now = datetime.now(timezone.utc)
    response = client.notify(
        event_type=EVENT,
        identifier={"experiment": "nbtest", "date": now.strftime("%Y%m%d"), "time": now.strftime("%H%M")},
        payload={"output_path": f"/tmp/aviso-notebook-test/{now:%Y%m%d%H%M%S}"},
    )
    print(response.status, response.request_id, response.processed_at)
else:
    print("skipped (set PUBLISH_TEST = True to publish)")
````

## 11. Error handling

Every failure is a subclass of `pyaviso.AvisoError`. `HttpError` carries the HTTP status,
the server's JSON body and the request id, which is what to quote when asking for help.

````python
cases = {
    "constraint on a string field": {"experiment": {"ne": "zzz"}},
    "missing the required field":   {"date": "20260101"},
    "unknown identifier key":       {"experiment": "*", "class": "d1"},
}
for label, bad_filter in cases.items():
    try:
        replay(client, EVENT, filter=bad_filter, start=1)
        print(f"{label}: accepted")
    except pyaviso.HttpError as exc:
        body = json.loads(exc.body) if exc.body else {}
        print(f"{label}: HTTP {exc.status} {body.get('code')}: {body.get('message')}")
    except pyaviso.AvisoError as exc:
        print(f"{label}: {type(exc).__name__}: {exc}")
````

````python
# Without a token the schema listing works, but replay and watch are refused.
anonymous = pyaviso.AvisoClient(base_url=BASE_URL)
print("anonymous schema listing:", sorted(anonymous.schema().event_types))
try:
    with anonymous.listen(EVENT, filter={"experiment": "*"}) as stream:
        next(iter(stream))
except pyaviso.HttpError as exc:
    print("anonymous watch: HTTP", exc.status, json.loads(exc.body).get("message"))
````

## 12. CLI equivalents

The wheel bundles the `aviso` command. It reads `~/.config/aviso/config.yaml`
(`base_url`, `auth.bearer_token`) by default; `--base-url` and `--token` (or the
`AVISO_BASE_URL` / `AVISO_TOKEN` env vars) override it. A separate `--state-file` keeps
these ad-hoc runs from touching your real resume cursors.

````python
!aviso --version
!aviso schema list --base-url {BASE_URL}
!aviso schema get {EVENT} --base-url {BASE_URL}
````

````python
# `| cat` because IPython runs `!` commands on a pseudo-terminal and the CLI only emits compact JSON when stdout is not a TTY.
!aviso replay --base-url {BASE_URL} --state-file output/cli-state.json --event {EVENT} --identifiers '{{"experiment":"*"}}' --from 1 --json | cat
````

Live watch with triggers is driven by a listener YAML file. Save this as
`de-scm-listener.yaml`:

```yaml
listeners:
  - name: de-scm-all
    event: de-scm-data
    identifiers:
      experiment: "*"
    triggers:
      - type: echo
      - type: log
        path: output/de-scm-data-live.ndjson
      - type: command
        command: "echo \"{{ notification.identifier.experiment }} -> {{ notification.payload.output_path }}\""
```

and run it in a terminal (Ctrl+C to stop):

```bash
aviso listen de-scm-listener.yaml
aviso listen de-scm-listener.yaml --from 2026-09-01     # replay from a publish date, then stay live
aviso replay de-scm-listener.yaml --from 1              # history only, then exit
```

`--from` accepts a sequence id (pure digits) or a UTC date such as `2026-09-01`,
`2026-09-01T06:00:00Z`. Pure digits are always a sequence id, never a date.
