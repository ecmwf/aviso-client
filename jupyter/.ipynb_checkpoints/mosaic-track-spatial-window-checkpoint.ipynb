# Spatial domain along the MOSAiC drift: Climate DT, CARRA2 and MOSAiC observations

A **moving 100 km × 100 km window** follows Polarstern along the MOSAiC drift track. For
every hour the window is centred on the ship position and filled from three sources, then
compared with the ship-based observations at the centre.

| Source key | Data | Native sampling in the window | Access |
| --- | --- | --- | --- |
| `climate-dt` | Destination Earth Climate DT, story-nudging IFS-FESOM (Gen 2, high resolution, `hist`) | HEALPix cells ~12.7 km, hourly (`sfc`), daily (`o2d`) | Polytope trajectory request on the MN5 data bridge, cached per day |
| `carra2` | Copernicus Arctic Regional Reanalysis (CARRA2, pan-Arctic) | 2.5 km Lambert grid, 3-hourly analyses | Local netCDF from the Climate Data Store (`~/Git/polarhack/data/carra2`), optional new retrievals |
| `mosaic` | MOSAiC observations: Met City tower and radiometers at the ship (point), ArcLeads MODIS lead maps (1 km, daily, Nov–Apr) | ship point + 1 km cells | Local files under `~/Git/polarhack/data` |

Window geometry: an azimuthal-equidistant tangent plane centred on the hourly ship position,
axes pointing geographic east (x) and north (y), cells kept when |x| ≤ 50 km and |y| ≤ 50 km.
Every source ends up in one long table (`source, param, time, x_km, y_km, value`), so the
viewer at the end switches between sources and parameters with two dropdowns and a time
slider.

> **Period.** The default period `2019-10-25 … 2019-11-05` is chosen so that every layer has
> data in the executed output: the local CARRA2 file covers October 2019 and ArcLeads starts
> on 1 November. Change `PERIOD` below to any part of the drift; Climate DT is fetched on
> demand and cached, CARRA2 needs the retrieval cell in section 5 for other months or
> variables.

Kernel: **Python (polarhack)** (`earthkit-data`, `polytope-client`, `cdsapi`, `plotly`,
`ipywidgets`, `anywidget`).

## 1. Configuration

```python
import json
import logging
import time
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import xarray as xr
from pyproj import Proj

warnings.filterwarnings("ignore")
for name in ("polytope", "polytope.api", "earthkit.data", "urllib3"):
    logging.getLogger(name).setLevel(logging.ERROR)

POLARHACK = Path.home() / "Git" / "polarhack"
PISCO = Path.home() / "Git" / "PISCO_notebooks"

TRACK_FILE = POLARHACK / "data" / "MOSAiC_combined" / "MOSAiC_track_1h.txt"
OBS_CANDIDATES = [POLARHACK / "jupyter" / "data" / "MOSAiC_obs_combined_1h_v2.nc",
                  PISCO / "Data" / "mosaic-examples" / "MOSAiC_obs_combined_1h_v2.nc"]
OBS_FILE = next(p for p in OBS_CANDIDATES if p.exists())
CARRA2_DIRS = [POLARHACK / "data" / "carra2"]
ARCLEADS_FILES = sorted((POLARHACK / "data" / "Pangaea").glob("*_ArcLeads.nc"))

CACHE = Path("data") / "spatial-window"          # next to this notebook
(CACHE / "climate-dt").mkdir(parents=True, exist_ok=True)
(CACHE / "carra2").mkdir(parents=True, exist_ok=True)
CARRA2_DIRS.append(CACHE / "carra2")

PERIOD = ("2019-10-25 00:00", "2019-11-05 23:00")   # inclusive, UTC
WINDOW_KM = 100.0
HALF_KM = WINDOW_KM / 2

CDT_EXPERIMENT = "hist"        # story-nudging: "cont", "hist" or "Tplus2.0K"
CDT_RESOLUTION = "high"        # "high" (~12.7 km cells) or "standard" (~50 km, one cell per window)
FETCH_CLIMATE_DT = True        # fetch missing Climate DT days from Polytope (needs ~/.polytopeapirc)
FETCH_CARRA2 = False           # run the CDS retrieval in section 5 (slow, queued, large files)

SOURCES = {"climate-dt": "Climate DT (IFS-FESOM story-nudging, " + CDT_EXPERIMENT + ")",
           "carra2": "CARRA2 reanalysis",
           "mosaic": "MOSAiC observations"}
SOURCE_COLORS = {"climate-dt": "#1f77b4", "carra2": "#ff7f0e", "mosaic": "#2ca02c"}
print("observations:", OBS_FILE)
print("ArcLeads files:", [p.name for p in ARCLEADS_FILES])
```

## 2. Ship track and window geometry

The hourly master track of Polarstern (PS122 legs 1–5) gives the window centre for every
hour. `to_window_xy` projects any latitude/longitude onto the tangent plane centred on the
ship, so the same code crops a HEALPix point cloud, a Lambert grid and a polar-stereographic
lead map.

```python
track = pd.read_csv(TRACK_FILE, sep=r"\s+")
track["time"] = pd.to_datetime(track["Datetime"].astype(str), format="%Y%m%d%H")
track = (track.rename(columns={"Latitude": "lat", "Longitude": "lon"})[["time", "lat", "lon"]]
         .drop_duplicates("time").set_index("time").sort_index())
track["lon"] = (track["lon"] + 180.0) % 360.0 - 180.0

sel = track.loc[PERIOD[0]:PERIOD[1]]
hours = sel.index
days = pd.DatetimeIndex(hours.normalize().unique())
print(f"full track: {track.index[0]} .. {track.index[-1]} ({len(track)} hourly positions)")
print(f"selected:   {hours[0]} .. {hours[-1]} ({len(hours)} hours, {len(days)} days)")
print(f"latitude {sel.lat.min():.2f}..{sel.lat.max():.2f} N, longitude {sel.lon.min():.2f}..{sel.lon.max():.2f} E")


def to_window_xy(lat, lon, lat0, lon0):
    # east/north offsets in km on the azimuthal-equidistant plane centred at (lat0, lon0)
    proj = Proj(proj="aeqd", lat_0=float(lat0), lon_0=float(lon0), datum="WGS84", units="m")
    x, y = proj(np.asarray(lon, dtype=float), np.asarray(lat, dtype=float))
    return x / 1e3, y / 1e3


def in_window(x_km, y_km):
    return (np.abs(x_km) <= HALF_KM) & (np.abs(y_km) <= HALF_KM)


def ship_at(t):
    row = track.loc[t]
    return float(row.lat), float(row.lon)


# self-check: drift distance per day tells how far the window moves
daily_pos = sel.resample("1D").first()
step_km = [np.hypot(*to_window_xy(b.lat, b.lon, a.lat, a.lon)) for a, b in zip(daily_pos.itertuples(), daily_pos.iloc[1:].itertuples())]
print(f"daily drift of the window centre: median {np.median(step_km):.1f} km, max {np.max(step_km):.1f} km")
```

## 3. Parameter registry

One canonical parameter name per physical quantity, with the variable name and unit
conversion for each source. Parameters without a spatial field in a source still get the ship
point value (`obs`) for the time-series comparison.

| Parameter | Climate DT | CARRA2 (CDS variable) | MOSAiC |
| --- | --- | --- | --- |
| `t2m` | `2t` | `t2m` (`2m_temperature`) | `temp_2m` (Met City 2 m) |
| `wspd10` | √(`10u`² + `10v`²) | `si10` (`10m_wind_speed`) | `wspd_vec_mean_10m` |
| `msl` | `msl` | `msl` (`mean_sea_level_pressure`) | `atmos_pressure_2m` (station pressure, not reduced) |
| `lwp` | `tclw` | `tclw` (`total_column_cloud_liquid_water`) | `lwp_hatpro` |
| `iwv` | `tcwv` | `tciwv` (`total_column_integrated_water_vapour`) | `iwv_hatpro_miracp` |
| `siconc` | `avg_siconc` (daily) | `siconc` (`sea_ice_area_fraction`) | – |
| `lead_fraction` | – | – | ArcLeads leads + open water per 5 km cell (daily) |
| `lead_map` | – | – | ArcLeads classes on the 1 km grid (daily) |

```python
K0 = 273.15
PARAMS = {
    "t2m": dict(label="2 m temperature", units="°C", cmap="RdBu_r",
                cdt=("2t", lambda v: v - K0), carra=(["t2m", "2t"], lambda v: v - K0),
                obs=("temp_2m", None)),
    "wspd10": dict(label="10 m wind speed", units="m s⁻¹", cmap="Viridis",
                   cdt=(("10u", "10v"), None), carra=(["si10", "10si", "wspd10"], None),
                   obs=("wspd_vec_mean_10m", None)),
    "msl": dict(label="Mean sea-level pressure", units="hPa", cmap="Cividis",
                cdt=("msl", lambda v: v / 100.0), carra=(["msl", "prmsl"], lambda v: v / 100.0),
                obs=("atmos_pressure_2m", None)),
    "lwp": dict(label="Liquid water path", units="g m⁻²", cmap="Blues",
                cdt=("tclw", lambda v: v * 1e3), carra=(["tclw"], lambda v: v * 1e3),
                obs=("lwp_hatpro", lambda v: v * 1e3)),
    "iwv": dict(label="Integrated water vapour", units="kg m⁻²", cmap="YlGnBu",
                cdt=("tcwv", None), carra=(["tciwv", "tcwv"], None),
                obs=("iwv_hatpro_miracp", None)),
    "siconc": dict(label="Sea-ice concentration", units="fraction", cmap="Blues_r",
                   cdt=("avg_siconc", None), carra=(["siconc", "ci", "sic"], None), obs=None),
    "lead_fraction": dict(label="Lead + open-water fraction (ArcLeads, 5 km cells)", units="fraction",
                          cmap="Reds", cdt=None, carra=None, obs=None),
    "lead_map": dict(label="ArcLeads class (0 cloud, 1 land, 2 sea ice, 3 artefact, 4 lead, 5 open water)",
                     units="class", cmap="Turbo", cdt=None, carra=None, obs=None),
}
DAILY_PARAMS = {"siconc", "lead_fraction", "lead_map"}   # one field per day, valid for all 24 hours

WINDOW_COLUMNS = ["source", "param", "time", "x_km", "y_km", "lat", "lon", "cell_km", "value"]
```

## 4. MOSAiC observations at the ship

The combined hourly file merges the Met City tower, the HATPRO/MiRAC-P radiometers and the
ASFS stations. These values sit at the window centre and are the reference for every
parameter that the ship measured.

```python
obs_ds = xr.open_dataset(OBS_FILE)
obs = pd.DataFrame(index=hours)
for p, spec in PARAMS.items():
    if spec.get("obs") is None:
        continue
    var, conv = spec["obs"]
    if var not in obs_ds:
        continue
    s = obs_ds[var].to_series()
    s = s[~s.index.duplicated()].reindex(hours)
    obs[p] = conv(s) if conv else s
print("observed parameters in the period:", {c: int(obs[c].notna().sum()) for c in obs.columns})
obs.describe().round(2)
```

## 5. Climate DT: Polytope trajectory retrieval around the track

One request per track day and parameter group. The day's 24 hourly positions are the
trajectory; every position is inflated to a box (`INFLATION_DEG` in latitude, widened by
1/cos φ in longitude so the box keeps its physical size at 85 °N). Polytope returns the
HEALPix cells inside the union of these boxes for all 24 hours as CoverageJSON. Each response
is saved to `data/spatial-window/climate-dt/`, so re-running only fetches missing days.

The box is requested larger than the window on purpose. The exact 100 km crop happens
locally in section 5.2 for every hour, so the drift within a day never pushes the window out
of the fetched area.

```python
import earthkit.data

earthkit.data.config.set("cache-policy", "off")

ADDRESS = "polytope.mn5.apps.dte.destination-earth.eu"   # story-nudging lives on the MN5 bridge
COLLECTION = "destination-earth"
INFLATION_DEG = 1.5            # half-size of the requested box in degrees latitude (see text)
RETRIES, RETRY_WAIT = 3, 15

BASE_REQUEST = {
    "class": "d1", "dataset": "climate-dt", "activity": "story-nudging",
    "experiment": CDT_EXPERIMENT, "generation": "2", "realization": "1",
    "model": "IFS-FESOM", "resolution": CDT_RESOLUTION, "expver": "0001",
    "stream": "clte", "type": "fc",
}
GROUPS = {
    "sfc": dict(levtype="sfc", param="2t/10u/10v/msl/tclw/tcwv",
                time="/".join(f"{h:02d}00" for h in range(24))),
    "o2d": dict(levtype="o2d", param="avg_siconc/avg_sithick", time="0000"),
}


def day_request(group, day):
    pts = track.loc[day:day + pd.Timedelta(hours=23)]
    lon_infl = float(INFLATION_DEG / np.cos(np.deg2rad(pts.lat.abs().max())))
    return {**BASE_REQUEST, **GROUPS[group], "date": f"{day:%Y%m%d}",
            "feature": {"type": "trajectory",
                        "points": pts[["lat", "lon"]].round(4).values.tolist(),
                        "inflation": [INFLATION_DEG, round(lon_infl, 3)],
                        "inflate": "box", "axes": ["latitude", "longitude"]}}


def tidy(ds):
    # CoverageJSON -> (time, points) with a datetime64 time axis
    ds = ds.drop_vars([c for c in ("levelist", "time") if c in ds.coords])
    ds = ds.squeeze([d for d in ("number", "steps") if d in ds.dims], drop=True)
    ds = ds.rename(datetimes="time")
    ds = ds.assign_coords(time=pd.to_datetime(ds.time.values, utc=True).tz_localize(None))
    return ds.sortby("time")


def chunk_path(group, day):
    return CACHE / "climate-dt" / f"{group}_{CDT_EXPERIMENT}_{CDT_RESOLUTION}_{day:%Y%m%d}.nc"


def fetch_day(group, day):
    path = chunk_path(group, day)
    if path.exists():
        return xr.open_dataset(path)
    if not FETCH_CLIMATE_DT:
        return None
    last = None
    for attempt in range(1, RETRIES + 1):
        try:
            src = earthkit.data.from_source("polytope", COLLECTION, day_request(group, day),
                                            address=ADDRESS, stream=False)
            ds = tidy(src.to_xarray())
            ds.attrs["inflation_deg"] = INFLATION_DEG
            ds.to_netcdf(path.with_suffix(".nc.part"))
            path.with_suffix(".nc.part").rename(path)
            return xr.open_dataset(path)
        except Exception as exc:     # network / server errors: retry, then give up on this day
            last = exc
            time.sleep(RETRY_WAIT)
    print(f"  {group} {day:%Y-%m-%d}: giving up after {RETRIES} attempts ({type(last).__name__}: {str(last)[:120]})")
    return None


t0 = time.time()
cdt_chunks = {g: {} for g in GROUPS}
for g in GROUPS:
    for day in days:
        ds = fetch_day(g, day)
        if ds is not None:
            cdt_chunks[g][day] = ds
    print(f"{g}: {len(cdt_chunks[g])}/{len(days)} days available")
print(f"done in {time.time() - t0:.0f} s; cached under {CACHE / 'climate-dt'}")
if cdt_chunks["sfc"]:
    sample = next(iter(cdt_chunks["sfc"].values()))
    print(f"sample day: {sample.sizes['points']} HEALPix points, variables {list(sample.data_vars)}")
```

### 5.1 Coverage check

For every hour the fetched point cloud must extend beyond ±50 km in both axes, otherwise the
window would be clipped. The cell also estimates the cell spacing from nearest-neighbour
distances, which sets the marker size in the maps.

```python
from scipy.spatial import cKDTree


def cell_spacing_km(lat, lon):
    if len(lat) < 2:
        return np.nan
    xyz = np.c_[np.cos(np.radians(lat)) * np.cos(np.radians(lon)),
                np.cos(np.radians(lat)) * np.sin(np.radians(lon)),
                np.sin(np.radians(lat))] * 6371.0
    dist, _ = cKDTree(xyz).query(xyz, k=2)
    return float(np.median(dist[:, 1]))


coverage = []
for day, ds in cdt_chunks["sfc"].items():
    for t in ds.time.values:
        t = pd.Timestamp(t)
        if t not in hours:
            continue
        lat0, lon0 = ship_at(t)
        x, y = to_window_xy(ds.latitude.values, ds.longitude.values, lat0, lon0)
        coverage.append(dict(time=t, xmin=x.min(), xmax=x.max(), ymin=y.min(), ymax=y.max(),
                             n_in_window=int(in_window(x, y).sum())))
coverage = pd.DataFrame(coverage).set_index("time")
CDT_CELL_KM = np.nanmedian([cell_spacing_km(ds.latitude.values, ds.longitude.values)
                            for ds in cdt_chunks["sfc"].values()]) if cdt_chunks["sfc"] else np.nan
ok = ((coverage.xmin <= -HALF_KM) & (coverage.xmax >= HALF_KM)
      & (coverage.ymin <= -HALF_KM) & (coverage.ymax >= HALF_KM))
print(f"Climate DT cell spacing: {CDT_CELL_KM:.1f} km -> about {int((WINDOW_KM / CDT_CELL_KM) ** 2)} cells per window")
print(f"window fully covered in {int(ok.sum())} of {len(ok)} hours; cells inside the window: "
      f"min {coverage.n_in_window.min()}, median {int(coverage.n_in_window.median())}, max {coverage.n_in_window.max()}")
if not ok.all():
    print("hours with clipped coverage (increase INFLATION_DEG):")
    display(coverage[~ok].head())
```

### 5.2 Crop the Climate DT fields to the moving window

Hourly `sfc` fields are matched to the ship position of the same hour; the daily `o2d`
fields (sea-ice concentration) are valid for all 24 hours of their day and use the position
at 12 UTC.

```python
def cdt_values(ds, param):
    spec = PARAMS[param]["cdt"]
    var, conv = spec
    if isinstance(var, tuple):                       # wind speed from components
        vals = np.hypot(ds[var[0]], ds[var[1]])
    else:
        if var not in ds:
            return None
        vals = ds[var]
    return conv(vals) if conv else vals


def window_rows_cdt():
    rows = []
    for group, chunks in cdt_chunks.items():
        for day, ds in chunks.items():
            params = [p for p, s in PARAMS.items() if s["cdt"] is not None
                      and (s["cdt"][0] in ds if not isinstance(s["cdt"][0], tuple) else s["cdt"][0][0] in ds)]
            if not params:
                continue
            lat, lon = ds.latitude.values, ds.longitude.values
            daily = group == "o2d"
            times = [day] if daily else [pd.Timestamp(t) for t in ds.time.values if pd.Timestamp(t) in hours]
            for t in times:
                if daily and not (hours.normalize() == day).any():
                    continue
                lat0, lon0 = ship_at(t + pd.Timedelta(hours=12) if daily else t)
                x, y = to_window_xy(lat, lon, lat0, lon0)
                m = in_window(x, y)
                if not m.any():
                    continue
                field = ds.isel(time=0) if daily else ds.sel(time=t)
                for p in params:
                    v = cdt_values(field, p)
                    if v is None:
                        continue
                    v = np.asarray(v.values, dtype=float)[m]
                    rows.append(pd.DataFrame({"source": "climate-dt", "param": p, "time": t,
                                              "x_km": x[m], "y_km": y[m], "lat": lat[m], "lon": lon[m],
                                              "cell_km": CDT_CELL_KM, "value": v}))
    return pd.concat(rows, ignore_index=True) if rows else pd.DataFrame(columns=WINDOW_COLUMNS)


win_cdt = window_rows_cdt()
print(f"{len(win_cdt):,} Climate DT window cells;", win_cdt.groupby("param").size().to_dict())
```

## 6. CARRA2: local pan-Arctic files cropped to the window

The Climate Data Store delivers CARRA2 on the full 2869 × 2869 Lambert grid (the `area`
keyword is not applied for this dataset), so each file is first cut to an index box around
the selected track and then cropped hour by hour. Every netCDF under
`~/Git/polarhack/data/carra2` and `data/spatial-window/carra2` is scanned; variables are
recognised by name, so files retrieved with the cell below join automatically.

```python
if FETCH_CARRA2:
    # Retrieve more CARRA2 variables or another month. Valid names come from the CDS form of
    # `reanalysis-pan-carra` (single_levels, analysis, 3-hourly). Expect a queue and ~250 MB
    # per variable-month; the area keyword is ignored by the CDS for this dataset.
    import cdsapi

    start, end = pd.Timestamp(PERIOD[0]), pd.Timestamp(PERIOD[1])
    for (year, month), sub in sel.groupby([sel.index.year, sel.index.month]):
        target = CACHE / "carra2" / f"carra2_sfc_{year}{month:02d}.nc"
        if target.exists():
            print("exists:", target)
            continue
        request = {
            "level_type": "single_levels",
            "variable": ["2m_temperature", "10m_wind_speed", "mean_sea_level_pressure",
                         "total_column_cloud_liquid_water", "total_column_integrated_water_vapour",
                         "sea_ice_area_fraction"],
            "product_type": "analysis",
            "time": ["00:00", "03:00", "06:00", "09:00", "12:00", "15:00", "18:00", "21:00"],
            "year": [str(year)], "month": [f"{month:02d}"],
            "day": [f"{d:02d}" for d in sorted(sub.index.day.unique())],
            "data_format": "netcdf",
        }
        cdsapi.Client().retrieve("reanalysis-pan-carra", request).download(str(target))
        print("saved", target)
else:
    print("FETCH_CARRA2 is False: using the CARRA2 files already on disk")
```

```python
def carra_param_for(varname):
    for p, spec in PARAMS.items():
        if spec.get("carra") and varname in spec["carra"][0]:
            return p
    return None


def window_rows_carra2():
    rows = []
    files = sorted(f for d in CARRA2_DIRS for f in d.glob("*.nc") if "ArcLeads" not in f.name)
    lat_margin, lon_margin = 1.0, 8.0          # degrees around the selected track
    for f in files:
        ds = xr.open_dataset(f)
        params = {v: carra_param_for(v) for v in ds.data_vars}
        params = {v: p for v, p in params.items() if p}
        if not params or "latitude" not in ds.coords:
            continue
        tname = "valid_time" if "valid_time" in ds.coords and ds["valid_time"].ndim == 1 else "time"
        times = pd.DatetimeIndex(ds[tname].values)
        keep = [i for i, t in enumerate(times) if t in hours]
        if not keep:
            continue
        lat2d = ds.latitude.values
        lon2d = (ds.longitude.values + 180.0) % 360.0 - 180.0
        box = ((lat2d >= sel.lat.min() - lat_margin) & (lat2d <= sel.lat.max() + lat_margin)
               & (lon2d >= sel.lon.min() - lon_margin) & (lon2d <= sel.lon.max() + lon_margin))
        jy, jx = np.where(box)
        ys, xs = slice(jy.min(), jy.max() + 1), slice(jx.min(), jx.max() + 1)
        ydim, xdim = ds.latitude.dims
        lat_c, lon_c = lat2d[ys, xs], lon2d[ys, xs]
        cell = float(np.hypot((lat_c[0, 1] - lat_c[0, 0]) * 111.2,
                              (lon_c[0, 1] - lon_c[0, 0]) * 111.2 * np.cos(np.radians(lat_c[0, 0]))))
        for var, p in params.items():
            sub = ds[var].isel({ydim: ys, xdim: xs}).isel({ds[var].dims[0]: keep}).load()
            conv = PARAMS[p]["carra"][1]
            for k, i in enumerate(keep):
                t = times[i]
                lat0, lon0 = ship_at(t)
                x, y = to_window_xy(lat_c.ravel(), lon_c.ravel(), lat0, lon0)
                m = in_window(x, y)
                v = np.asarray(sub.isel({sub.dims[0]: k}).values, dtype=float).ravel()[m]
                v = conv(v) if conv else v
                rows.append(pd.DataFrame({"source": "carra2", "param": p, "time": t,
                                          "x_km": x[m], "y_km": y[m], "lat": lat_c.ravel()[m],
                                          "lon": lon_c.ravel()[m], "cell_km": cell, "value": v}))
        print(f"{f.name}: {list(params.values())}, {len(keep)} analysis times in the period, cell {cell:.2f} km")
    return pd.concat(rows, ignore_index=True) if rows else pd.DataFrame(columns=WINDOW_COLUMNS)


win_carra = window_rows_carra2()
print(f"{len(win_carra):,} CARRA2 window cells;", win_carra.groupby("param").size().to_dict())
```

## 7. MOSAiC spatial observations: ArcLeads lead maps in the window

ArcLeads (Willmes et al.) classifies MODIS thermal-infrared imagery into daily 1 km maps
(0 cloud, 1 land, 2 sea ice, 3 artefact, 4 lead, 5 open water) for November to April. Two
layers are derived per day and window:

- `lead_map`: the raw classes on the 1 km grid;
- `lead_fraction`: (leads + open water) / (sea ice + leads + open water) on 5 km cells, left
  empty where clouds or artefacts hide more than half of a cell.

Both are daily and centred on the ship position at 12 UTC.

```python
def window_rows_arcleads(agg_km=5.0):
    rows = []
    for f in ARCLEADS_FILES:
        ds = xr.open_dataset(f)
        ftimes = pd.to_datetime(ds.time.values.astype(str), format="%Y%m%d")
        idx = [i for i, t in enumerate(ftimes) if t in days]
        if not idx:
            continue
        lat2d, lon2d = ds.lat.values, ds.lon.values
        for i in idx:
            day = ftimes[i]
            lat0, lon0 = ship_at(day + pd.Timedelta(hours=12))
            # coarse pre-selection in degrees, then exact projection on the small crop
            dlat = np.abs(lat2d - lat0) * 111.2
            dlon = (np.abs(lon2d - lon0) % 360.0)
            dlon = np.where(dlon > 180.0, 360.0 - dlon, dlon) * 111.2 * np.cos(np.radians(lat0))
            jy, jx = np.where((dlat <= HALF_KM * 1.3) & (dlon <= HALF_KM * 1.3))
            ys, xs = slice(jy.min(), jy.max() + 1), slice(jx.min(), jx.max() + 1)
            classes = ds.leadmap.isel(time_dim=i, nrows=ys, ncols=xs).values.astype(float)
            lat_c, lon_c = lat2d[ys, xs], lon2d[ys, xs]
            x, y = to_window_xy(lat_c.ravel(), lon_c.ravel(), lat0, lon0)
            m = in_window(x, y)
            cell = float(np.hypot(x[1] - x[0], y[1] - y[0]))
            rows.append(pd.DataFrame({"source": "mosaic", "param": "lead_map", "time": day,
                                      "x_km": x[m], "y_km": y[m], "lat": lat_c.ravel()[m],
                                      "lon": lon_c.ravel()[m], "cell_km": cell, "value": classes.ravel()[m]}))
            # aggregate to agg_km cells
            c = classes.ravel()[m]
            gx = np.floor((x[m] + HALF_KM) / agg_km).astype(int)
            gy = np.floor((y[m] + HALF_KM) / agg_km).astype(int)
            frame = pd.DataFrame({"gx": gx, "gy": gy,
                                  "valid": np.isin(c, [2, 4, 5]), "open": np.isin(c, [4, 5])})
            g = frame.groupby(["gx", "gy"]).agg(n=("valid", "size"), valid=("valid", "sum"), open=("open", "sum"))
            frac = np.where(g.valid >= 0.5 * g.n, g.open / g.valid.clip(lower=1), np.nan)
            cx = (g.index.get_level_values("gx") + 0.5) * agg_km - HALF_KM
            cy = (g.index.get_level_values("gy") + 0.5) * agg_km - HALF_KM
            rows.append(pd.DataFrame({"source": "mosaic", "param": "lead_fraction", "time": day,
                                      "x_km": cx, "y_km": cy, "lat": np.nan, "lon": np.nan,
                                      "cell_km": agg_km, "value": frac}))
        print(f"{f.name}: {len(idx)} days in the period")
    return pd.concat(rows, ignore_index=True) if rows else pd.DataFrame(columns=WINDOW_COLUMNS)


win_leads = window_rows_arcleads()
print(f"{len(win_leads):,} ArcLeads window cells;", win_leads.groupby("param").size().to_dict())
```

## 8. One table for everything

All window cells of all sources are stacked into `win`; `stats` holds the window mean,
spread and extremes per source, parameter and time together with the ship observation at
that time. Both are cached as compressed CSV next to the notebook.

```python
win = pd.concat([win_cdt, win_carra, win_leads], ignore_index=True)
win["time"] = pd.to_datetime(win["time"])
win = win.dropna(subset=["value"]).reset_index(drop=True)

stats = (win.groupby(["source", "param", "time"])["value"]
         .agg(mean="mean", std="std", min="min", max="max", n="size").reset_index())
obs_long = obs.stack().rename("obs").reset_index()
obs_long.columns = ["time", "param", "obs"]
stats = stats.merge(obs_long, on=["time", "param"], how="left")
stats["obs"] = np.where(stats.param.isin(DAILY_PARAMS), np.nan, stats["obs"])

win.to_csv(CACHE / "window_cells.csv.gz", index=False)
stats.to_csv(CACHE / "window_stats.csv.gz", index=False)
print(f"{len(win):,} window cells, {len(stats):,} (source, param, time) summaries")
print("\nparameters available per source:")
avail = win.groupby(["source", "param"]).agg(times=("time", "nunique"), cells_per_time=("time", lambda s: int(len(s) / s.nunique())))
avail
```

## 9. Interactive viewer

Two dropdowns choose the **source** and **parameter**, the slider moves the window along
the track. For 3-hourly (CARRA2) and daily (ArcLeads, sea ice) layers the nearest available
field is shown and its valid time is written in the title. The right-hand panel shows the
window mean (line) and ±1 spatial standard deviation (band) of the selected source together
with the other sources' means and the ship observation; the vertical line marks the slider
time.

Options: *fixed colour range* uses the 2–98 % range of the parameter over all sources and
times so switching sources keeps the colours comparable; *show track* draws the ±24 h track
inside the window.

```python
import contextlib

import ipywidgets as widgets
import plotly.graph_objects as go
from plotly.subplots import make_subplots

_nullcontext = contextlib.nullcontext

try:
    import anywidget  # noqa: F401  (needed by go.FigureWidget for in-place updates)
    HAVE_FIGUREWIDGET = True
except ImportError:
    HAVE_FIGUREWIDGET = False

TIMES = pd.DatetimeIndex(sorted(win.time.unique()))
HOURLY = [t for t in hours]
PX_PER_KM = 4.2            # map panel is ~420 px wide for 100 km

# fixed colour range per parameter (2..98 percentile across all sources and times)
CRANGE = {p: (float(g.value.quantile(0.02)), float(g.value.quantile(0.98))) for p, g in win.groupby("param")}
CRANGE["lead_map"] = (0, 5)
CRANGE["lead_fraction"] = (0, max(0.05, CRANGE.get("lead_fraction", (0, 0.2))[1]))
CRANGE["siconc"] = (min(0.8, CRANGE.get("siconc", (0.8, 1))[0]), 1.0)


def params_for(source):
    return sorted(win.loc[win.source == source, "param"].unique(), key=list(PARAMS).index)


def nearest_field_time(source, param, t):
    times = win.loc[(win.source == source) & (win.param == param), "time"].unique()
    if len(times) == 0:
        return None
    times = pd.DatetimeIndex(sorted(times))
    if param in DAILY_PARAMS:
        same_day = times[times.normalize() == pd.Timestamp(t).normalize()]
        return same_day[0] if len(same_day) else None
    i = times.get_indexer([pd.Timestamp(t)], method="nearest")[0]
    return times[i] if abs(times[i] - pd.Timestamp(t)) <= pd.Timedelta(hours=1.5) else None


def field(source, param, t):
    return win[(win.source == source) & (win.param == param) & (win.time == t)]


def track_segment(t, hours_before=24, hours_after=24):
    lat0, lon0 = ship_at(t)
    seg = track.loc[pd.Timestamp(t) - pd.Timedelta(hours=hours_before): pd.Timestamp(t) + pd.Timedelta(hours=hours_after)]
    x, y = to_window_xy(seg.lat.values, seg.lon.values, lat0, lon0)
    return x, y


def build_figure():
    fig = make_subplots(rows=1, cols=2, column_widths=[0.42, 0.58], horizontal_spacing=0.08,
                        subplot_titles=("window", "window statistics along the drift"))
    fig.add_trace(go.Scattergl(x=[], y=[], mode="markers", name="field", showlegend=False,
                               marker=dict(symbol="square", size=6, colorscale="Viridis", showscale=True,
                                           colorbar=dict(x=0.40, len=0.9, thickness=14, title=""))), row=1, col=1)
    fig.add_trace(go.Scatter(x=[], y=[], mode="lines", name="track ±24 h", line=dict(color="black", width=1.5),
                             showlegend=False), row=1, col=1)
    fig.add_trace(go.Scatter(x=[0], y=[0], mode="markers", name="Polarstern", showlegend=False,
                             marker=dict(symbol="star", size=14, color="white", line=dict(color="black", width=1.5))),
                  row=1, col=1)
    # time-series panel: band (2 traces), selected mean, other means, observation
    fig.add_trace(go.Scatter(x=[], y=[], mode="lines", line=dict(width=0), showlegend=False, hoverinfo="skip"), row=1, col=2)
    fig.add_trace(go.Scatter(x=[], y=[], mode="lines", line=dict(width=0), fill="tonexty",
                             fillcolor="rgba(120,120,120,0.25)", name="±1 σ (window)", hoverinfo="skip"), row=1, col=2)
    for s in SOURCES:
        fig.add_trace(go.Scatter(x=[], y=[], mode="lines", name=f"{s} window mean",
                                 line=dict(color=SOURCE_COLORS[s], width=1.2)), row=1, col=2)
    fig.add_trace(go.Scatter(x=[], y=[], mode="lines", name="MOSAiC obs at ship",
                             line=dict(color="black", width=1.2, dash="dot")), row=1, col=2)
    fig.update_xaxes(title_text="east (km)", range=[-HALF_KM, HALF_KM], constrain="domain", row=1, col=1)
    fig.update_yaxes(title_text="north (km)", range=[-HALF_KM, HALF_KM], scaleanchor="x", scaleratio=1, row=1, col=1)
    fig.update_xaxes(title_text="time (UTC)", row=1, col=2)
    fig.update_layout(height=520, width=1050, margin=dict(l=50, r=20, t=70, b=50),
                      legend=dict(orientation="h", y=-0.18, x=0.4), template="plotly_white")
    fig.add_shape(type="line", x0=0, x1=0, y0=0, y1=1, xref="x2", yref="paper",
                  line=dict(color="crimson", width=1.5, dash="dash"), name="cursor")
    return fig


def fill_figure(fig, source, param, t, fixed_range=True, show_track=True):
    t = pd.Timestamp(t)
    ft = nearest_field_time(source, param, t)
    spec = PARAMS[param]
    df = field(source, param, ft) if ft is not None else win.iloc[0:0]
    size = float(df.cell_km.iloc[0]) * PX_PER_KM * 0.98 if len(df) else 6
    lo, hi = CRANGE[param] if fixed_range else ((float(df.value.min()), float(df.value.max())) if len(df) else (0, 1))
    seg_x, seg_y = track_segment(t) if show_track else ([], [])
    st = stats[(stats.source == source) & (stats.param == param)].sort_values("time")
    with (fig.batch_update() if HAVE_FIGUREWIDGET else _nullcontext()):
        fig.data[0].update(x=df.x_km, y=df.y_km,
                           marker=dict(color=df.value, cmin=lo, cmax=hi, size=max(size, 2.5),
                                       colorscale=spec["cmap"], symbol="square", showscale=True,
                                       colorbar=dict(x=0.40, len=0.9, thickness=14, title=dict(text=spec["units"]))),
                           hovertemplate="x %{x:.1f} km, y %{y:.1f} km<br>%{marker.color:.3g} " + spec["units"] + "<extra></extra>")
        fig.data[1].update(x=seg_x, y=seg_y)
        fig.data[3].update(x=st.time, y=st["mean"] - st["std"])
        fig.data[4].update(x=st.time, y=st["mean"] + st["std"])
        for k, s in enumerate(SOURCES):
            ss = stats[(stats.source == s) & (stats.param == param)].sort_values("time")
            fig.data[5 + k].update(x=ss.time, y=ss["mean"], visible=True if len(ss) else "legendonly")
        o = obs[param] if param in obs else pd.Series(dtype=float)
        fig.data[8].update(x=o.index, y=o.values, visible=True if o.notna().any() else "legendonly")
        fig.layout.shapes[0].update(x0=t, x1=t)
        valid = f"{ft:%Y-%m-%d %H:%M}" if ft is not None else "no field within tolerance"
        ncell = len(df)
        stat_txt = (f"mean {df.value.mean():.3g}, σ {df.value.std():.3g}, min {df.value.min():.3g}, max {df.value.max():.3g} {spec['units']}"
                    if ncell else "")
        obs_val = obs.at[t, param] if (param in obs and t in obs.index) else np.nan
        obs_txt = f" | ship obs {obs_val:.3g} {spec['units']}" if np.isfinite(obs_val) else ""
        lat0, lon0 = ship_at(t)
        fig.layout.title.text = (f"<b>{SOURCES[source]}</b> — {spec['label']}<br>"
                                 f"<sup>window {WINDOW_KM:.0f}×{WINDOW_KM:.0f} km at {lat0:.2f}°N {lon0:.2f}°E, "
                                 f"{t:%Y-%m-%d %H:%M} UTC | field valid {valid} | {ncell} cells: {stat_txt}{obs_txt}</sup>")
        fig.layout.yaxis2.title.text = f"{spec['label']} ({spec['units']})"
    return fig


# ── widgets ────────────────────────────────────────────────────────
w_source = widgets.Dropdown(options=[(SOURCES[s], s) for s in SOURCES if s in win.source.unique()],
                            description="source", layout=widgets.Layout(width="360px"))
w_param = widgets.Dropdown(options=[(PARAMS[p]["label"], p) for p in params_for(w_source.value)],
                           description="parameter", layout=widgets.Layout(width="420px"))
w_time = widgets.SelectionSlider(options=[(f"{t:%Y-%m-%d %H:%M}", t) for t in HOURLY], description="time",
                                 continuous_update=False, layout=widgets.Layout(width="1000px"))
w_fixed = widgets.Checkbox(value=True, description="fixed colour range", indent=False)
w_track = widgets.Checkbox(value=True, description="show track ±24 h", indent=False)
w_play = widgets.Play(interval=600, min=0, max=len(HOURLY) - 1, step=3, description="play")
widgets.jslink((w_play, "value"), (w_time, "index"))

fig = go.FigureWidget(build_figure()) if HAVE_FIGUREWIDGET else build_figure()
out = widgets.Output()


def on_source_change(change):
    opts = [(PARAMS[p]["label"], p) for p in params_for(change["new"])]
    current = w_param.value
    w_param.options = opts
    w_param.value = current if current in [p for _, p in opts] else opts[0][1]


def redraw(*_):
    fill_figure(fig, w_source.value, w_param.value, w_time.value, w_fixed.value, w_track.value)
    if not HAVE_FIGUREWIDGET:
        with out:
            out.clear_output(wait=True)
            fig.show()


w_source.observe(on_source_change, names="value")
for w in (w_source, w_param, w_time, w_fixed, w_track):
    w.observe(redraw, names="value")

# start at midday of the first day that has both a CARRA2 and a Climate DT field
start_t = next((t for t in HOURLY if t.hour == 12), HOURLY[0])
w_time.value = start_t
redraw()
controls = widgets.VBox([widgets.HBox([w_source, w_param, w_fixed, w_track]), widgets.HBox([w_play, w_time])])
display(widgets.VBox([controls, fig if HAVE_FIGUREWIDGET else out]))
```

### 9.1 All sources side by side

Same parameter, same slider time, one window per source. Useful to see how the ~12.7 km
Climate DT cells, the 2.5 km CARRA2 grid and the 1 km lead map resolve the same area.

```python
def side_by_side(param, t, fixed_range=True):
    t = pd.Timestamp(t)
    sources = [s for s in SOURCES if len(win[(win.source == s) & (win.param == param)])]
    if param.startswith("lead"):
        sources = ["mosaic"]
    fig = make_subplots(rows=1, cols=max(len(sources), 1), horizontal_spacing=0.06,
                        subplot_titles=[SOURCES[s] for s in sources])
    lo, hi = CRANGE[param]
    for k, s in enumerate(sources, start=1):
        ft = nearest_field_time(s, param, t)
        df = field(s, param, ft) if ft is not None else win.iloc[0:0]
        if not fixed_range and len(df):
            lo, hi = float(df.value.min()), float(df.value.max())
        size = float(df.cell_km.iloc[0]) * 3.0 * 0.98 if len(df) else 6
        fig.add_trace(go.Scattergl(x=df.x_km, y=df.y_km, mode="markers", showlegend=False,
                                   marker=dict(symbol="square", size=max(size, 2.5), color=df.value, cmin=lo, cmax=hi,
                                               colorscale=PARAMS[param]["cmap"], showscale=(k == len(sources)),
                                               colorbar=dict(title=PARAMS[param]["units"], thickness=14)),
                                   hovertemplate="%{marker.color:.3g}<extra>" + s + "</extra>"), row=1, col=k)
        sx, sy = track_segment(t)
        fig.add_trace(go.Scatter(x=sx, y=sy, mode="lines", line=dict(color="black", width=1), showlegend=False), row=1, col=k)
        fig.add_trace(go.Scatter(x=[0], y=[0], mode="markers", showlegend=False,
                                 marker=dict(symbol="star", size=12, color="white", line=dict(color="black", width=1.5))), row=1, col=k)
        valid = f"valid {ft:%Y-%m-%d %H:%M}" if ft is not None else "no field"
        fig.layout.annotations[k - 1].text = f"{SOURCES[s]}<br><sup>{valid}, {len(df)} cells</sup>"
        fig.update_xaxes(range=[-HALF_KM, HALF_KM], title_text="east (km)", row=1, col=k)
        fig.update_yaxes(range=[-HALF_KM, HALF_KM], scaleanchor=f"x{k if k > 1 else ''}", title_text="north (km)" if k == 1 else None, row=1, col=k)
    lat0, lon0 = ship_at(t)
    fig.update_layout(height=380, width=330 * max(len(sources), 1) + 80, template="plotly_white",
                      title=f"{PARAMS[param]['label']} — {t:%Y-%m-%d %H:%M} UTC, window at {lat0:.2f}°N {lon0:.2f}°E",
                      margin=dict(t=90, b=50))
    return fig


w2_param = widgets.Dropdown(options=[(PARAMS[p]["label"], p) for p in sorted(win.param.unique(), key=list(PARAMS).index)],
                            description="parameter", layout=widgets.Layout(width="420px"))
w2_time = widgets.SelectionSlider(options=[(f"{t:%Y-%m-%d %H:%M}", t) for t in HOURLY], description="time",
                                  value=start_t, continuous_update=False, layout=widgets.Layout(width="1000px"))
out2 = widgets.Output()


def redraw2(*_):
    with out2:
        out2.clear_output(wait=True)
        side_by_side(w2_param.value, w2_time.value).show()


w2_param.observe(redraw2, names="value")
w2_time.observe(redraw2, names="value")
redraw2()
display(widgets.VBox([widgets.HBox([w2_param]), w2_time, out2]))
```

### 9.2 Kernel-free version

The two viewers above need a running kernel. This figure carries the data for one parameter

still works in an exported HTML file or a static notebook viewer. Change `STATIC_PARAM` for

```python
STATIC_PARAM = "t2m"
step_times = [t for t in HOURLY if t.hour % 3 == 0]
sources_static = [s for s in SOURCES if len(win[(win.source == s) & (win.param == STATIC_PARAM)])]
lo, hi = CRANGE[STATIC_PARAM]

fig_s = go.Figure()
for s in sources_static:
    fig_s.add_trace(go.Scattergl(x=[], y=[], mode="markers", name=SOURCES[s], visible=(s == sources_static[0]),
                                 marker=dict(symbol="square", size=6, color=[], cmin=lo, cmax=hi,
                                             colorscale=PARAMS[STATIC_PARAM]["cmap"], showscale=True,
                                             colorbar=dict(title=PARAMS[STATIC_PARAM]["units"]))))
fig_s.add_trace(go.Scatter(x=[0], y=[0], mode="markers", showlegend=False, name="Polarstern",
                           marker=dict(symbol="star", size=14, color="white", line=dict(color="black", width=1.5))))

steps = []
for t in step_times:
    xs, ys, cs, sizes = [], [], [], []
    for s in sources_static:
        ft = nearest_field_time(s, STATIC_PARAM, t)
        df = field(s, STATIC_PARAM, ft) if ft is not None else win.iloc[0:0]
        xs.append(df.x_km.round(2).tolist()); ys.append(df.y_km.round(2).tolist())
        cs.append(df.value.round(3).tolist())
        sizes.append(max(float(df.cell_km.iloc[0]) * 4.6, 2.5) if len(df) else 6)
    lat0, lon0 = ship_at(t)
    steps.append(dict(method="update", label=f"{t:%m-%d %H}h",
                      args=[{"x": xs + [[0]], "y": ys + [[0]], "marker.color": cs + [None], "marker.size": sizes + [14]},
                            {"title.text": f"{PARAMS[STATIC_PARAM]['label']} — {t:%Y-%m-%d %H:%M} UTC, {lat0:.2f}°N {lon0:.2f}°E"}]))

buttons = [dict(label=SOURCES[s], method="update",
                args=[{"visible": [s2 == s for s2 in sources_static] + [True]}]) for s in sources_static]
fig_s.update_layout(
    sliders=[dict(steps=steps, active=0, currentvalue=dict(prefix="time: "), pad=dict(t=40))],
    updatemenus=[dict(buttons=buttons, direction="down", x=0.0, xanchor="left", y=1.18, yanchor="top")],
    xaxis=dict(title="east (km)", range=[-HALF_KM, HALF_KM]),
    yaxis=dict(title="north (km)", range=[-HALF_KM, HALF_KM], scaleanchor="x"),
    width=620, height=640, template="plotly_white", showlegend=False,
    title=f"{PARAMS[STATIC_PARAM]['label']} — use the dropdown for the source and the slider for time")
# initialise with the first step
first = steps[0]["args"][0]
for k, tr in enumerate(fig_s.data):
    tr.x, tr.y = first["x"][k], first["y"][k]
    if first["marker.color"][k] is not None:
        tr.marker.color = first["marker.color"][k]
    tr.marker.size = first["marker.size"][k]
fig_s.show()
```

## 10. Window statistics versus the ship observations

`stats` already holds the window mean per source and time next to the observation at the
ship. The first table compares each source with the observations on the times where both
exist; the second summarises the *spatial* variability inside the window (mean σ over the
period), which tells how representative a single point is for the 100 km domain.

```python
def score(g):
    d = g.dropna(subset=["mean", "obs"])
    if len(d) < 3:
        return pd.Series(dict(n=len(d), bias=np.nan, rmsd=np.nan, r=np.nan))
    diff = d["mean"] - d["obs"]
    return pd.Series(dict(n=len(d), bias=diff.mean(), rmsd=np.sqrt((diff ** 2).mean()), r=d["mean"].corr(d["obs"])))


scores = (stats[~stats.param.isin(DAILY_PARAMS)].groupby(["param", "source"]).apply(score, include_groups=False)
          .dropna(subset=["bias"]).round(3))
scores.columns = ["n (times)", "bias (window mean − obs)", "RMSD", "r"]
display(scores)

spatial = (stats.groupby(["param", "source"]).agg(mean_sigma=("std", "mean"), max_sigma=("std", "max"),
                                                  mean_range=("max", lambda s: np.nan), n_times=("time", "nunique")))
spatial["mean_range"] = stats.assign(rng=stats["max"] - stats["min"]).groupby(["param", "source"])["rng"].mean()
spatial.columns = ["mean σ in window", "max σ in window", "mean (max − min) in window", "n times"]
display(spatial.round(3))
```

```python
def rgba(hex_color, alpha):
    r, g, b = (int(hex_color[i:i + 2], 16) for i in (1, 3, 5))
    return f"rgba({r},{g},{b},{alpha})"


p = "t2m"
st = stats[stats.param == p].sort_values("time")
fig_ts = go.Figure()
for s in SOURCES:
    ss = st[st.source == s]
    if not len(ss):
        continue
    fig_ts.add_trace(go.Scatter(x=pd.concat([ss.time, ss.time[::-1]]), y=pd.concat([ss["mean"] + ss["std"], (ss["mean"] - ss["std"])[::-1]]),
                                fill="toself", fillcolor=rgba(SOURCE_COLORS[s], 0.18), line=dict(width=0),
                                showlegend=False, hoverinfo="skip"))
    fig_ts.add_trace(go.Scatter(x=ss.time, y=ss["mean"], mode="lines", name=f"{s} window mean ± σ",
                                line=dict(color=SOURCE_COLORS[s], width=1.4)))
if p in obs:
    fig_ts.add_trace(go.Scatter(x=obs.index, y=obs[p], mode="lines", name="MOSAiC obs at ship", line=dict(color="black", width=1.2, dash="dot")))
fig_ts.update_layout(template="plotly_white", height=380, width=1000,
                     title=f"{PARAMS[p]['label']}: window mean ± spatial σ per source, and the ship observation",
                     yaxis_title=f"{PARAMS[p]['label']} ({PARAMS[p]['units']})", legend=dict(orientation="h", y=-0.25))
fig_ts.show()
```

## Notes and caveats

- **Story-nudging** keeps the Climate DT large-scale circulation close to the reanalysis of the
  actual day, so a day-by-day comparison with observations is meaningful. The free-running
  `scenariomip` and `baseline` experiments would not be comparable in time. Change
  `CDT_EXPERIMENT` to `cont` or `Tplus2.0K` for the storyline counterfactuals (they use the
  same request shape).
- Polytope's trajectory `inflation` was measured to behave like a box *width* rather than a
  half-width at these latitudes, which is why the request uses 1.5° and the coverage check in
  5.1 exists. The exact 100 km crop is always done locally.
- CARRA2 `msl` versus the MOSAiC `atmos_pressure_2m` are not the same quantity: the ship
  value is station pressure at 2 m and is not reduced to sea level (the difference is small
  at sea level but systematic).
- ArcLeads is a satellite product, not a MOSAiC in-situ measurement, and the classes are
  sensitive to clouds (class 0) and thin-ice detection. PANGAEA also publishes lead fractions
  within a 50 km radius around Polarstern from seven sensors (von Albedyll et al. 2024,
  <https://doi.org/10.1594/PANGAEA.963736>). The download is protected by a browser
  challenge, so it cannot be scripted here; downloaded CSVs dropped into
  `data/spatial-window/` can be merged with `stats` on the day.
- Retrieval notes: the DESP token in `~/.polytopeapirc` comes from
  `~/Git/polarhack/desp-authentication.py`; run it again on HTTP 401. CDS retrievals of CARRA2
  are queued and deliver the full pan-Arctic grid.
- Cached artefacts under `data/spatial-window/`: one netCDF per Climate DT day and group,
  `window_cells.csv.gz` (all cells) and `window_stats.csv.gz` (summaries). Delete them to
  force a rebuild.
