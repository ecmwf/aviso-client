# Aviso 2 on the LUMI data bridge: Extremes-DT events → earthkit retrieval → map

This notebook connects the **Aviso 2 client** (`pyaviso` 2.x) to the Destination Earth
notification server on the LUMI data bridge, waits for new **Extremes-DT data-availability
events**, and for every event it builds a **Polytope request with earthkit-data**, retrieves the
field and **plots it with earthkit-plots**. Drop-down menus let you change the Aviso filter
(stream, level type, step) and the retrieval (parameter, level, map domain) without editing code.

| Setting | Value |
| --- | --- |
| Aviso server | `https://aviso2.lumi.apps.dte.destination-earth.eu` (override with `AVISO_BASE_URL`) |
| Event type | `data` (MARS-like identifiers: `class`, `stream`, `type`, `levtype`, `date`, `time`, `step`, ...) |
| Aviso auth | Bearer token: your **DESP offline token** (`~/.polytopeapirc`, written by `desp-authentication.py`), `AVISO_TOKEN`, or an ECMWF API key |
| Data | Polytope at `polytope.lumi.apps.dte.destination-earth.eu`, dataset `extremes-dt` (DESP credentials in `~/.polytopeapirc`) |
| Client | `pip install pyaviso earthkit-data earthkit-plots ipywidgets pyyaml pandas` |

> **Hostname.** The address `aviso2.lumi.apps.destination-earth.eu` (without `.dte.`) does not
> resolve in DNS. The Aviso 2 server answers at `aviso2.lumi.apps.dte.destination-earth.eu`.
> The neighbouring `aviso.lumi.apps.dte.destination-earth.eu` (no `2`) is still the Aviso 1.x
> server and needs `pyaviso==1.0.2`; see `../DE/aviso-examples/`.

Sections:

1. Setup and authentication
2. Discover the `data` schema
3. Replay the recent history
4. Request builder: drop-down menus
5. Event → earthkit retrieval → map (dry run on the latest event)
6. Listen live and process events as they arrive
7. Resume across restarts
8. CLI equivalent and notes

## 1. Setup and authentication

`AVISO_BASE_URL` and `AVISO_TOKEN` in the environment take precedence. Without `AVISO_TOKEN`
the notebook tries the DESP offline token from `~/.polytopeapirc` (the LUMI server is fronted by
DESP's Keycloak, so the same token that Polytope uses is accepted) and then the ECMWF key in
`~/.config/aviso/config.yaml`. Each candidate is probed with a tiny replay; the first one the
server accepts is used. No token is ever printed.

````python
# Uncomment to install into the active kernel:
# %pip install -q pyaviso earthkit-data earthkit-plots ipywidgets pyyaml pandas

import json
import os
import pathlib
import threading
import time
from datetime import datetime, timedelta, timezone

import ipywidgets as widgets
import pandas as pd
import pyaviso
import earthkit.data as ekd
import earthkit.plots as ekp
from IPython.display import display

print("pyaviso", pyaviso.__version__, "| earthkit-data", ekd.__version__, "| earthkit-plots", ekp.__version__)
````

````python
BASE_URL = os.environ.get("AVISO_BASE_URL", "https://aviso2.lumi.apps.dte.destination-earth.eu").rstrip("/")
EVENT = "data"                                   # the only event type on this server
POLYTOPE_ADDRESS = "polytope.lumi.apps.dte.destination-earth.eu"
DATASET = "extremes-dt"
OUT_DIR = pathlib.Path("output") / "lumi-extremes-dt"
OUT_DIR.mkdir(parents=True, exist_ok=True)

CLI_CONFIG = pathlib.Path.home() / ".config" / "aviso" / "config.yaml"
POLYTOPE_RC = pathlib.Path.home() / ".polytopeapirc"


def token_candidates():
    # Ordered (label, token) pairs; secrets stay in memory only.
    out = []
    if os.environ.get("AVISO_TOKEN"):
        out.append(("AVISO_TOKEN", os.environ["AVISO_TOKEN"]))
    if POLYTOPE_RC.exists():
        key = json.loads(POLYTOPE_RC.read_text(encoding="utf-8")).get("user_key")
        if key:
            out.append(("DESP offline token (~/.polytopeapirc)", key))
    if CLI_CONFIG.exists():
        import yaml
        cfg = yaml.safe_load(CLI_CONFIG.read_text(encoding="utf-8")) or {}
        key = (cfg.get("auth") or {}).get("bearer_token")
        if key:
            out.append(("ECMWF key (~/.config/aviso/config.yaml)", key))
    return out


def open_stream(client, event_type, *, filter, start=None, mode=None, triggers=None):
    # pyaviso 2.1.0 calls the cursor `start_from=`; newer builds call it `from_=`.
    kwargs = {"filter": filter}
    if mode is not None:
        kwargs["mode"] = mode
    if triggers is not None:
        kwargs["triggers"] = triggers
    if start is None:
        return client.listen(event_type, **kwargs)
    try:
        return client.listen(event_type, start_from=start, **kwargs)
    except TypeError:
        return client.listen(event_type, from_=start, **kwargs)


def connect(base_url=BASE_URL, **client_kwargs):
    # Try each token with a one-notification replay; keep the first the server accepts.
    tried = []
    for label, tok in token_candidates():
        client = pyaviso.AvisoClient(base_url=base_url, auth=pyaviso.Bearer(tok), **client_kwargs)
        try:
            with open_stream(client, EVENT, filter={"class": "d1"}, start=1, mode="replay_only") as s:
                next(iter(s), None)
            print(f"connected to {base_url} using {label}")
            return client, tok
        except pyaviso.HttpError as exc:
            tried.append(f"{label}: HTTP {exc.status}")
        except pyaviso.AvisoError as exc:
            tried.append(f"{label}: {type(exc).__name__}: {exc}")
    raise RuntimeError(
        "No accepted token for " + base_url + ". Tried: " + ("; ".join(tried) or "nothing found") +
        ". Run ../DE/aviso-examples/desp-authentication.py to refresh ~/.polytopeapirc or set AVISO_TOKEN."
    )


client, TOKEN = connect()
````

## 2. Discover the `data` schema

The server owns the schema; the client discovers it. Only `class` is **required** in a watch or
replay filter, every other identifier is optional and narrows the match. The payload is optional
on this stream (LUMI notifications usually carry none), so the identifiers *are* the information:
they tell you which forecast field is now available.

````python
catalog = client.schema()
print("event types:", catalog.event_types)
schema = client.schema_for(EVENT).schema
rows = []
for key, spec in schema["identifier"].items():
    rows.append({"identifier": key, "type": spec.get("type"), "required": spec.get("required", False),
                 "values / format": spec.get("values") or spec.get("canonical_format") or spec.get("range") or spec.get("default") or ""})
pd.DataFrame(rows).set_index("identifier")
````

## 3. Replay the recent history

`mode="replay_only"` streams what the server retains and stops at `replay_completed`. A
timestamp cursor selects by **publish time** on the server, not by the forecast `date`. The
frame below is also used to fill the drop-down menus with the values that actually occur.

````python
def replay(client, *, filter, start=1):
    with open_stream(client, EVENT, filter=filter, start=start, mode="replay_only") as stream:
        return list(stream)


def to_frame(notifications):
    rows = []
    for n in notifications:
        ce = n.cloudevent or {}
        row = {"sequence": n.sequence, "published_utc": ce.get("time")}
        row.update(n.identifier)
        row["payload"] = n.payload
        rows.append(row)
    frame = pd.DataFrame(rows)
    if not frame.empty:
        frame["published_utc"] = pd.to_datetime(frame["published_utc"], utc=True, format="ISO8601")
    return frame


REPLAY_HOURS = 48
since = (datetime.now(timezone.utc) - timedelta(hours=REPLAY_HOURS)).strftime("%Y-%m-%dT%H:%M:%SZ")
history = replay(client, filter={"class": "d1"}, start=since)
df = to_frame(history)
print(f"{len(history)} notifications published since {since}")
df.tail(10)
````

````python
if not df.empty:
    summary = (df.groupby(["stream", "type", "levtype"], dropna=False)
                 .agg(events=("sequence", "size"),
                      first_date=("date", "min"), last_date=("date", "max"),
                      steps=("step", lambda s: f"{min(s)}..{max(s)} ({s.nunique()} distinct)"),
                      last_published=("published_utc", "max")))
    display(summary)
````

## 4. Request builder: drop-down menus

Two groups of controls:

* **Aviso filter** (left): which notifications you want to be told about. `class` is fixed to
  `d1`; `stream`, `type`, `levtype` and `step` are read from the schema and the history. `any`
  leaves the key out of the filter.
* **Retrieval and plot** (right): what to fetch from Polytope for each event and how to draw it.
  The `param` list follows the chosen `stream`/`levtype`; `levelist` appears for `pl`/`hl`.

Change a menu at any time. The Aviso filter is read when a listen starts (section 6); the
retrieval settings are read **per event**, so you can switch parameter or domain while listening.

````python
def schema_values(key, fallback):
    spec = schema["identifier"].get(key, {})
    return list(spec.get("values") or fallback)

# --- parameter catalogue for the Extremes-DT (paramId strings as Polytope expects them)
PARAMS = {
    ("oper", "sfc"): {"2t: 2 metre temperature": "167", "10u: 10 m U wind": "165", "10v: 10 m V wind": "166",
                      "msl: mean sea level pressure": "151", "tp: total precipitation": "228",
                      "2d: 2 metre dewpoint": "168", "skt: skin temperature": "235", "sp: surface pressure": "134",
                      "tcc: total cloud cover": "164", "ssrd: surface solar radiation downwards": "169",
                      "100u: 100 m U wind": "228246", "100v: 100 m V wind": "228247"},
    ("oper", "pl"):  {"t: temperature": "130", "u: U wind": "131", "v: V wind": "132", "q: specific humidity": "133",
                      "z: geopotential": "129", "r: relative humidity": "157", "w: vertical velocity": "135"},
    ("oper", "hl"):  {"t: temperature": "130", "u: U wind": "131", "v: V wind": "132", "q: specific humidity": "133"},
    ("wave", "sfc"): {"swh: significant wave height": "140229", "mwp: mean wave period": "140232",
                      "mwd: mean wave direction": "140230", "pp1d: peak wave period": "140231"},
}
LEVELS = {"pl": ["1000", "925", "850", "700", "500", "300", "250", "200", "100", "50"], "hl": ["100", "500", "1000"]}
DOMAINS = ["Europe", "Global", "Central Europe", "Northwest Europe", "Northeast Europe", "Southwest Europe",
           "Southeast Europe", "Mediterranean", "North Atlantic", "Africa", "Arctic", "North America"]

steps_seen = sorted(int(s) for s in df["step"].dropna().unique()) if not df.empty else []
step_options = ["any"] + (steps_seen or [0, 3, 6, 12, 24, 48, 72, 96, 120])

# --- Aviso filter controls
w_stream = widgets.Dropdown(options=schema_values("stream", ["oper", "wave"]), value="oper", description="stream")
w_type = widgets.Dropdown(options=["any", "fc"], value="fc", description="type")
w_levtype = widgets.Dropdown(options=schema_values("levtype", ["sfc", "pl", "hl"]), value="sfc", description="levtype")
w_step = widgets.Dropdown(options=step_options, value=step_options[0], description="step")

# --- retrieval / plot controls
w_param = widgets.Dropdown(description="param")
w_levelist = widgets.Dropdown(description="levelist")
w_domain = widgets.Dropdown(options=DOMAINS, value="Europe", description="domain")
w_save = widgets.Checkbox(value=True, description="save GRIB + PNG to output/")


def refresh_param_options(*_):
    key = (w_stream.value, w_levtype.value if w_stream.value == "oper" else "sfc")
    opts = PARAMS.get(key) or PARAMS[("oper", "sfc")]
    current = w_param.value
    w_param.options = opts
    w_param.value = current if current in opts.values() else next(iter(opts.values()))
    lev = LEVELS.get(w_levtype.value)
    w_levelist.options = lev or ["-"]
    w_levelist.value = (lev or ["-"])[0]
    w_levelist.layout.visibility = "visible" if lev else "hidden"


w_stream.observe(refresh_param_options, names="value")
w_levtype.observe(refresh_param_options, names="value")
refresh_param_options()

display(widgets.HBox([
    widgets.VBox([widgets.HTML("<b>Aviso filter</b> (class=d1)"), w_stream, w_type, w_levtype, w_step]),
    widgets.VBox([widgets.HTML("<b>Retrieval and plot</b>"), w_param, w_levelist, w_domain, w_save]),
]))
````

````python
def aviso_filter():
    # Filter dict for client.listen(); "any" drops the key.
    f = {"class": "d1", "stream": w_stream.value, "levtype": w_levtype.value}
    if w_type.value != "any":
        f["type"] = w_type.value
    if w_step.value != "any":
        f["step"] = int(w_step.value)
    return f


def build_request(notification):
    # Map the notification identifiers onto a Polytope/MARS request and add the menu choices.
    ident = notification.identifier
    req = {
        "class": ident.get("class", "d1"),
        "dataset": DATASET,
        "expver": str(ident.get("expver", "0001")),
        "stream": ident.get("stream", w_stream.value),
        "type": ident.get("type", "fc"),
        "levtype": ident.get("levtype", w_levtype.value),
        "date": str(ident["date"]),
        "time": str(ident["time"]),
        "step": str(ident["step"]),
        "param": w_param.value,
    }
    if req["levtype"] in LEVELS:
        req["levelist"] = w_levelist.value
    return req


print("current Aviso filter:", aviso_filter())
````

## 5. Event → earthkit retrieval → map

`process(notification)` is the whole pipeline: build the request, fetch with
`earthkit.data.from_source("polytope", ...)`, plot with `earthkit.plots.Map`. A single
Extremes-DT surface field is a global reduced Gaussian grid of ~26 million points (about 50 MB
of GRIB, roughly 20 s to download here); earthkit-plots resamples it for the chosen domain.

The button runs the pipeline on the **latest retained notification** that matches the current
filter, so you can check the whole chain before waiting for a new event.

````python
def retrieve(request):
    data = ekd.from_source("polytope", "destination-earth", request, address=POLYTOPE_ADDRESS, stream=False)
    fields = data.to_fieldlist() if hasattr(data, "to_fieldlist") else data
    if w_save.value:
        stem = "{stream}_{levtype}_{param}_{date}_{time}_step{step}".format(**request)
        fields.to_target("file", str(OUT_DIR / f"{stem}.grib"))
    return fields


def visualise(fields, request, domain=None):
    domain = domain or w_domain.value
    chart = ekp.Map(domain=None if domain == "Global" else domain)
    chart.quickplot(fields[0])
    chart.coastlines()
    chart.gridlines()
    chart.legend()
    chart.title("Extremes-DT {variable_name}, valid {valid_time:%Y-%m-%d %H:%M} (T+{step})")
    if w_save.value:
        stem = "{stream}_{levtype}_{param}_{date}_{time}_step{step}".format(**request)
        chart.save(OUT_DIR / f"{stem}_{domain.replace(' ', '')}.png")
    chart.show()
    return chart


def process(notification):
    request = build_request(notification)
    print(f"seq={notification.sequence} identifier={notification.identifier}")
    print("polytope request:", json.dumps(request))
    t0 = time.monotonic()
    fields = retrieve(request)
    print(f"retrieved {len(fields)} field(s) in {time.monotonic() - t0:.1f}s")
    display(fields.ls())
    visualise(fields, request)
    return fields


out = widgets.Output()
btn = widgets.Button(description="Retrieve & plot latest matching event", button_style="primary",
                     layout=widgets.Layout(width="320px"))


def on_click(_=None):
    with out:
        out.clear_output(wait=True)
        f = aviso_filter()
        recent = replay(client, filter=f, start=since)
        if not recent:
            print("no notification in the last", REPLAY_HOURS, "hours matches", f)
            return
        process(recent[-1])


btn.on_click(on_click)
display(btn, out)
````

````python
# Run the dry run once on execution so the notebook shows a result without a click.
RUN_DRY_RUN_NOW = True
if RUN_DRY_RUN_NOW:
    on_click()
````

## 6. Listen live and process events as they arrive

`client.listen(EVENT, filter=...)` without a mode is a live watch: it blocks until the next
matching notification and reconnects on its own. The helper closes the stream from a timer
thread after `seconds` or after `max_events`, whichever comes first, so the cell always returns.

* `start=` replays from a cursor (sequence id or publish-time ISO timestamp) and then stays
  live. Use it to catch up after a pause. Leave it `None` to only see new events.
* Every event that passes the Aviso filter is handed to `process(...)`, which reads the
  **current** parameter / level / domain menus, so you can change those while it runs.
* The Extremes-DT publishes many steps per run. Set `step` in the filter (or a small
  `max_events`) unless you want a retrieval per step.

````python
def listen_and_process(client, *, seconds=300, max_events=None, start=None, handler=process):
    f = aviso_filter()
    print(f"listening on {client.base_url} for {EVENT} {f} (up to {seconds}s"
          f"{', max ' + str(max_events) + ' event(s)' if max_events else ''}"
          f"{', from ' + str(start) if start is not None else ''})")
    stream = open_stream(client, EVENT, filter=f, start=start)
    timer = threading.Timer(seconds, stream.close)
    timer.start()
    handled = []
    t0 = time.monotonic()
    try:
        for n in stream:
            print(f"\n[{time.monotonic() - t0:6.1f}s] event seq={n.sequence} published={(n.cloudevent or {}).get('time')}")
            try:
                handler(n)
                handled.append(n)
            except Exception as exc:                # keep listening if one retrieval fails
                print(f"  processing failed: {type(exc).__name__}: {exc}")
            if max_events and len(handled) >= max_events:
                break
    finally:
        timer.cancel()
        stream.close()
    print(f"\nwatch closed after {time.monotonic() - t0:.1f}s, {len(handled)} event(s) processed")
    return handled


# Live only, short window so the notebook keeps moving; raise `seconds` (e.g. 6 * 3600) for a real session.
LISTEN_SECONDS = 20
live = listen_and_process(client, seconds=LISTEN_SECONDS, max_events=1)
````

````python
# Catch-up variant: replay everything published in the last 2 hours that matches the filter, then stay live.
# Left disabled by default because each matching event triggers a retrieval.
CATCH_UP = False
if CATCH_UP:
    cursor = (datetime.now(timezone.utc) - timedelta(hours=2)).strftime("%Y-%m-%dT%H:%M:%SZ")
    caught_up = listen_and_process(client, seconds=600, max_events=3, start=cursor)
````

## 7. Resume across restarts

With a `JsonFileStore` the client commits the last delivered sequence per (server, event,
filter). Restarting the notebook with the same filter resumes after that cursor, so events
published while the kernel was down are still processed. `flush_cursor_on_exit=True` commits
the final position when the stream is closed.

````python
state_path = OUT_DIR / "state.json"
resuming = pyaviso.AvisoClient(
    base_url=BASE_URL,
    auth=pyaviso.Bearer(TOKEN),
    state_store=pyaviso.JsonFileStore(state_path),
    flush_cursor_on_exit=True,
)
# First run: no cursor yet, so start from the replay window; later runs continue from the stored cursor.
resumed = listen_and_process(resuming, seconds=20, max_events=1, start=since)
if state_path.exists():
    print(json.dumps(json.loads(state_path.read_text(encoding="utf-8")), indent=2))
````

## 8. CLI equivalent and notes

The wheel ships the `aviso` command. The same watch, with a `command` trigger that calls a
Python script for each event, runs unattended in a terminal:

```yaml
# lumi-extremes-dt.yaml
listeners:
  - name: extremes-dt-sfc-step0
    event: data
    identifiers:
      class: d1
      stream: oper
      type: fc
      levtype: sfc
      step: 0
    triggers:
      - type: echo
      - type: log
        path: output/lumi-extremes-dt/events.ndjson
      - type: command
        command: >-
          python retrieve_and_plot.py --date {{ notification.identifier.date }}
          --time {{ notification.identifier.time }} --step {{ notification.identifier.step }}
        timeout: 600
```

```bash
export AVISO_BASE_URL=https://aviso2.lumi.apps.dte.destination-earth.eu
export AVISO_TOKEN="$(python -c 'import json,pathlib;print(json.load(open(pathlib.Path.home()/".polytopeapirc"))["user_key"])')"
aviso schema list
aviso listen lumi-extremes-dt.yaml --from 2026-09-24T00:00:00Z    # catch up from a publish time, then live
```

Notes:

* **401 on watch/replay** means the token was rejected. Refresh the DESP offline token with
  `python ../DE/aviso-examples/desp-authentication.py` (it rewrites `~/.polytopeapirc`).
  The schema listing works without a token.
* **Polytope errors** (`403`, "not found") come from the data side, not from Aviso. Check the
  request printed by `process()`; steps and levels not produced by the Extremes-DT return an
  empty result.
* **Volume.** One surface field is ~50 MB. Filter on `step` or keep `max_events` small while
  experimenting; use `earthkit-regrid` if you need coarser fields for further processing.
