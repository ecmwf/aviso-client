# Spatial domain along the MOSAiC drift: Climate DT, CARRA2 and MOSAiC observations

A **moving 100 km × 100 km window** follows Polarstern along the MOSAiC drift track. For
every hour the window is centred on the ship position and filled from three sources, then
compared with the ship-based observations at the centre.

| Dataset key | Data | Native sampling | Access |
| --- | --- | --- | --- |
| `climate-dt` | Destination Earth Climate DT, story-nudging IFS-FESOM (Gen 2, high resolution, `hist`) | HEALPix nside 512 cells, 12.7 km, hourly (`sfc`), daily (`o2d`) | Polytope on the DestinE MN5 data bridge, DESP token from `desp-authentication.py`, cached per day |
| `carra2` | Copernicus Arctic Regional Reanalysis (CARRA2, pan-Arctic) | 2.5 km Lambert grid, 3-hourly analyses | Local netCDF from the Climate Data Store (`~/Git/polarhack/data/carra2`), optional new retrievals |
| `mosaic` | MOSAiC observations: Met City tower and radiometers at the ship (point), ArcLeads MODIS lead maps (1 km, daily, Nov–Apr) | ship point + 1 km cells | Local files under `~/Git/polarhack/data` |

Two time ranges are used throughout:

- **`PERIOD`** (default 25 Oct – 5 Nov 2019): the days shown on the maps and in the window
  viewers. Chosen so that every layer has data: the local CARRA2 file covers October 2019,
  ArcLeads starts on 1 November.
- **the full track** (20 Sep 2019 – 12 Oct 2020): the window statistics in section 10 and the
  time-series panels cover the whole drift for every dataset that has data there. Climate DT
  is fetched for all 389 days (one request per day and group, cached; the first run takes a
  few hours), CARRA2 and ArcLeads for whatever is on disk.

Window geometry: an azimuthal-equidistant tangent plane centred on the hourly ship position,
axes pointing geographic east (x) and north (y), cells kept when |x| ≤ 50 km and |y| ≤ 50 km.

Kernel: **Python (polarhack)** (`earthkit-data`, `polytope-client`, `cdsapi`, `healpy`,
`plotly`, `ipywidgets`, `anywidget`).

### 1.1 DestinE authentication

Climate DT is served by Polytope on the DestinE Service Platform and needs a DESP token in
`~/.polytopeapirc`. `desp-authentication.py` (from the polytope-examples repository, copy in
`~/Git/polarhack`) logs in with the DESP username and password and stores a long-lived
offline token. The cell checks the current token with a cheap Polytope call and only runs the
script when that check fails; it then prompts for the password (and the username unless
`DESP_USER` is set).

```python
import getpass
import subprocess
import sys

from polytope.api import Client as PolytopeClient

ADDRESS = "polytope.mn5.apps.dte.destination-earth.eu"   # story-nudging lives on the MN5 bridge
COLLECTION = "destination-earth"
TOKEN_FILE = Path.home() / ".polytopeapirc"


def polytope_token_ok():
    try:
        PolytopeClient(address=ADDRESS, quiet=True).list_collections()
        return True
    except Exception as exc:          # 401 for a missing or expired token, network errors otherwise
        print(f"Polytope token check failed: {type(exc).__name__}: {str(exc)[:160]}")
        return False


def desp_login():
    user = DESP_USER or input("DESP username: ")
    password = getpass.getpass("DESP password: ")
    # desp-authentication.py reads USER / PASSWORD from the environment (or -u / -p); the
    # environment keeps the password out of the process list.
    env = {**os.environ, "USER": user, "PASSWORD": password}
    subprocess.run([sys.executable, str(DESP_AUTH_SCRIPT), "-o", str(TOKEN_FILE)], env=env, check=True)


if FETCH_CLIMATE_DT:
    if polytope_token_ok():
        print(f"DESP token in {TOKEN_FILE} is valid")
    else:
        desp_login()
        if not polytope_token_ok():
            raise RuntimeError("Polytope still rejects the token after desp-authentication.py")
```

## 2. Ship track and window geometry

The hourly master track of Polarstern (PS122 legs 1–5) gives the window centre for every
hour. `to_window_xy` projects any latitude/longitude onto the tangent plane centred on the
ship, so the same code crops a HEALPix point cloud, a Lambert grid and a polar-stereographic
lead map; `from_window_xy` is its inverse and puts window corners and cells back on the map.

```python
track = pd.read_csv(TRACK_FILE, sep=r"\s+")
track["time"] = pd.to_datetime(track["Datetime"].astype(str), format="%Y%m%d%H")
track = (track.rename(columns={"Latitude": "lat", "Longitude": "lon"})[["time", "lat", "lon"]]
         .drop_duplicates("time").set_index("time").sort_index())
track["lon"] = (track["lon"] + 180.0) % 360.0 - 180.0

sel = track.loc[PERIOD[0]:PERIOD[1]]                       # display period
hours = sel.index
days = pd.DatetimeIndex(hours.normalize().unique())
stats_hours = track.index if STATS_FULL_TRACK else hours    # statistics range
stats_days = pd.DatetimeIndex(stats_hours.normalize().unique())
print(f"full track: {track.index[0]} .. {track.index[-1]} ({len(track)} hourly positions)")
print(f"display:    {hours[0]} .. {hours[-1]} ({len(hours)} hours, {len(days)} days), "
      f"{sel.lat.min():.2f}..{sel.lat.max():.2f} N, {sel.lon.min():.2f}..{sel.lon.max():.2f} E")
print(f"statistics: {stats_hours[0]} .. {stats_hours[-1]} ({len(stats_days)} days)")


def to_window_xy(lat, lon, lat0, lon0):
    # east/north offsets in km on the azimuthal-equidistant plane centred at (lat0, lon0)
    proj = Proj(proj="aeqd", lat_0=float(lat0), lon_0=float(lon0), datum="WGS84", units="m")
    x, y = proj(np.asarray(lon, dtype=float), np.asarray(lat, dtype=float))
    return x / 1e3, y / 1e3


def from_window_xy(x_km, y_km, lat0, lon0):
    proj = Proj(proj="aeqd", lat_0=float(lat0), lon_0=float(lon0), datum="WGS84", units="m")
    lon, lat = proj(np.asarray(x_km, dtype=float) * 1e3, np.asarray(y_km, dtype=float) * 1e3, inverse=True)
    return lat, lon


def in_window(x_km, y_km):
    return (np.abs(x_km) <= HALF_KM) & (np.abs(y_km) <= HALF_KM)


def ship_at(t):
    row = track.loc[t]
    return float(row.lat), float(row.lon)


def window_ring(t, n_per_side=10):
    # closed ring of the window edge in lat/lon (several points per side so it stays square on the map)
    lat0, lon0 = ship_at(t)
    s_ = np.linspace(-HALF_KM, HALF_KM, n_per_side)
    xs = np.r_[s_, np.full(n_per_side, HALF_KM), s_[::-1], np.full(n_per_side, -HALF_KM)]
    ys = np.r_[np.full(n_per_side, -HALF_KM), s_, np.full(n_per_side, HALF_KM), s_[::-1]]
    lat, lon = from_window_xy(xs, ys, lat0, lon0)
    return np.r_[lat, lat[0]], np.r_[lon, lon[0]]


# fixed map area for section 11: the display-period track plus a margin, or MAP_AREA
if MAP_AREA is None:
    dlat = MAP_MARGIN_KM / 111.2
    dlon = MAP_MARGIN_KM / (111.2 * np.cos(np.radians(sel.lat.mean())))
    MAP_AREA = (round(min(sel.lat.max() + dlat, 89.9), 1), round(sel.lon.min() - dlon, 1),
                round(sel.lat.min() - dlat, 1), round(sel.lon.max() + dlon, 1))
AREA_N, AREA_W, AREA_S, AREA_E = MAP_AREA
AREA_LAT0, AREA_LON0 = (AREA_N + AREA_S) / 2, (AREA_W + AREA_E) / 2
print(f"map area: {AREA_S}..{AREA_N} N, {AREA_W}..{AREA_E} E")
```

```python
K0 = 273.15
PARAMS = {
    "t2m": dict(label="2 m temperature", units="°C", cmap="RdBu_r",
                cdt=("2t", lambda v: v - K0), carra=(["t2m", "2t"], lambda v: v - K0),
                obs=("temp_2m", None)),
    "wspd10": dict(label="10 m wind speed", units="m s⁻¹", cmap="Viridis",
                   cdt=(("10u", "10v"), None), carra=(["si10", "10si", "wspd10"], None),
                   obs=("wspd_vec_mean_10m", None)),
    "msl": dict(label="Mean sea-level pressure", units="hPa", cmap="Cividis",
                cdt=("msl", lambda v: v / 100.0), carra=(["msl", "prmsl"], lambda v: v / 100.0),
                obs=("atmos_pressure_2m", None)),
    "lwp": dict(label="Liquid water path", units="g m⁻²", cmap="Blues",
                cdt=("tclw", lambda v: v * 1e3), carra=(["tclw"], lambda v: v * 1e3),
                obs=("lwp_hatpro", lambda v: v * 1e3)),
    "iwv": dict(label="Integrated water vapour", units="kg m⁻²", cmap="YlGnBu",
                cdt=("tcwv", None), carra=(["tciwv", "tcwv"], None),
                obs=("iwv_hatpro_miracp", None)),
    "siconc": dict(label="Sea-ice concentration", units="fraction", cmap="Blues_r",
                   cdt=("avg_siconc", None), carra=(["siconc", "ci", "sic"], None), obs=None),
    "lead_fraction": dict(label="Lead + open-water fraction (ArcLeads, 5 km cells)", units="fraction",
                          cmap="Reds", cdt=None, carra=None, obs=None),
    "lead_map": dict(label="ArcLeads class (0 cloud, 1 land, 2 sea ice, 3 artefact, 4 lead, 5 open water)",
                     units="class", cmap="Turbo", cdt=None, carra=None, obs=None),
}
DAILY_PARAMS = {"siconc", "lead_fraction", "lead_map"}   # one field per day, valid for all 24 hours
PARAM_ORDER = list(PARAMS)

WINDOW_COLUMNS = ["source", "param", "time", "x_km", "y_km", "lat", "lon", "cell_km", "value"]
STAT_COLUMNS = ["source", "param", "time", "mean", "std", "min", "max", "n"]


def summarise(values):
    v = np.asarray(values, dtype=float)
    v = v[np.isfinite(v)]
    if v.size == 0:
        return None
    return dict(mean=float(v.mean()), std=float(v.std()), min=float(v.min()), max=float(v.max()), n=int(v.size))
```

## 5. Climate DT from Polytope

Two kinds of request, both cached under `data/spatial-window/climate-dt/` so re-running only
fetches what is missing:

- **Window chunks**, one per day and parameter group, for every day in the statistics range:
  a trajectory request along the day's 24 positions, each inflated to a box. Polytope treats
  the trajectory point as the south-west corner of its box, so the positions are shifted by
  half the inflation to centre the box on the ship. The box is larger than the window; the
  exact 100 km crop is done locally in 5.2.
- **Area chunks**, one per day and group, for the display period: a bounding-box request over
  the fixed map area (`MAP_AREA`). These give the complete, gap-free field on the map in
  section 11.

Parameter groups: `sfc` (2t, 10u, 10v, msl, tclw, tcwv, all 24 hours) and `o2d`
(avg_siconc, avg_sithick, daily at 00 UTC).

```python
import earthkit.data

earthkit.data.config.set("cache-policy", "off")

INFLATION_DEG = 1.5            # window request: box size in degrees latitude (width, not half-width)
RETRIES, RETRY_WAIT = 3, 15

BASE_REQUEST = {
    "class": "d1", "dataset": "climate-dt", "activity": "story-nudging",
    "experiment": CDT_EXPERIMENT, "generation": "2", "realization": "1",
    "model": "IFS-FESOM", "resolution": CDT_RESOLUTION, "expver": "0001",
    "stream": "clte", "type": "fc",
}
GROUPS = {
    "sfc": dict(levtype="sfc", param="2t/10u/10v/msl/tclw/tcwv",
                time="/".join(f"{h:02d}00" for h in range(24))),
    "o2d": dict(levtype="o2d", param="avg_siconc/avg_sithick", time="0000"),
}
AREA_TAG = f"{AREA_N}_{AREA_W}_{AREA_S}_{AREA_E}".replace("-", "m")


def window_request(group, day):
    pts = track.loc[day:day + pd.Timedelta(hours=23)]
    lon_infl = float(INFLATION_DEG / np.cos(np.deg2rad(pts.lat.abs().max())))
    corners = pts[["lat", "lon"]] - [INFLATION_DEG / 2, lon_infl / 2]
    return {**BASE_REQUEST, **GROUPS[group], "date": f"{day:%Y%m%d}",
            "feature": {"type": "trajectory", "points": corners.round(4).values.tolist(),
                        "inflation": [INFLATION_DEG, round(lon_infl, 3)],
                        "inflate": "box", "axes": ["latitude", "longitude"]}}


def area_request(group, day):
    return {**BASE_REQUEST, **GROUPS[group], "date": f"{day:%Y%m%d}",
            "feature": {"type": "boundingbox", "points": [[AREA_N, AREA_W], [AREA_S, AREA_E]]}}


def tidy(ds):
    # CoverageJSON -> (time, points) with a datetime64 time axis
    ds = ds.drop_vars([c for c in ("levelist", "time") if c in ds.coords])
    ds = ds.squeeze([d for d in ("number", "steps") if d in ds.dims], drop=True)
    ds = ds.rename(datetimes="time")
    ds = ds.assign_coords(time=pd.to_datetime(ds.time.values, utc=True).tz_localize(None))
    return ds.sortby("time")


def chunk_path(kind, group, day):
    if kind == "window":
        return CACHE / "climate-dt" / f"{group}_{CDT_EXPERIMENT}_{CDT_RESOLUTION}_{day:%Y%m%d}.nc"
    return CACHE / "climate-dt" / f"area_{group}_{CDT_EXPERIMENT}_{CDT_RESOLUTION}_{AREA_TAG}_{day:%Y%m%d}.nc"


def fetch_chunk(kind, group, day):
    path = chunk_path(kind, group, day)
    if path.exists():
        return path
    if not FETCH_CLIMATE_DT:
        return None
    request = (window_request if kind == "window" else area_request)(group, day)
    last = None
    for attempt in range(1, RETRIES + 1):
        try:
            src = earthkit.data.from_source("polytope", COLLECTION, request, address=ADDRESS, stream=False, quiet=True)
            ds = tidy(src.to_xarray())
            ds.attrs["request_kind"] = kind
            ds.to_netcdf(path.with_suffix(".nc.part"))
            path.with_suffix(".nc.part").rename(path)
            return path
        except Exception as exc:     # network / server errors: retry, then give up on this day
            last = exc
            time.sleep(RETRY_WAIT)
    print(f"  {kind} {group} {day:%Y-%m-%d}: giving up after {RETRIES} attempts ({type(last).__name__}: {str(last)[:120]})")
    return None


t0 = time.time()
missing_before = sum(not chunk_path("window", g, d).exists() for g in GROUPS for d in stats_days)
print(f"window chunks: {len(stats_days)} days x {len(GROUPS)} groups, {missing_before} to fetch")
window_files = {g: {} for g in GROUPS}
for k, day in enumerate(stats_days, 1):
    for g in GROUPS:
        path = fetch_chunk("window", g, day)
        if path is not None:
            window_files[g][day] = path
    if missing_before and k % 30 == 0:
        print(f"  {k}/{len(stats_days)} days, {time.time() - t0:.0f} s")
for g in GROUPS:
    print(f"  {g}: {len(window_files[g])}/{len(stats_days)} days on disk")

area_files = {g: {} for g in GROUPS}
for day in days:
    for g in GROUPS:
        path = fetch_chunk("area", g, day)
        if path is not None:
            area_files[g][day] = path
for g in GROUPS:
    print(f"area chunks {g}: {len(area_files[g])}/{len(days)} days on disk")
print(f"done in {time.time() - t0:.0f} s")
```

### 5.1 Coverage check (display period)

For every hour the window chunk must extend beyond ±50 km in both axes, and the area chunk
must cover the map area. The cell also measures the HEALPix cell spacing.

```python
from scipy.spatial import cKDTree


def cell_spacing_km(lat, lon):
    if len(lat) < 2:
        return np.nan
    xyz = np.c_[np.cos(np.radians(lat)) * np.cos(np.radians(lon)),
                np.cos(np.radians(lat)) * np.sin(np.radians(lon)), np.sin(np.radians(lat))] * 6371.0
    dist, _ = cKDTree(xyz).query(xyz, k=2)
    return float(np.median(dist[:, 1]))


coverage, spacing = [], []
for day in days:
    if day not in window_files["sfc"]:
        continue
    with xr.open_dataset(window_files["sfc"][day]) as ds:
        lat, lon = ds.latitude.values, ds.longitude.values
        times = [pd.Timestamp(t) for t in ds.time.values]
    spacing.append(cell_spacing_km(lat, lon))
    for t in times:
        if t not in hours:
            continue
        x, y = to_window_xy(lat, lon, *ship_at(t))
        coverage.append(dict(time=t, xmin=x.min(), xmax=x.max(), ymin=y.min(), ymax=y.max(), n_in_window=int(in_window(x, y).sum())))
coverage = pd.DataFrame(coverage).set_index("time")
CDT_CELL_KM = float(np.nanmedian(spacing)) if spacing else np.nan
ok = ((coverage.xmin <= -HALF_KM) & (coverage.xmax >= HALF_KM) & (coverage.ymin <= -HALF_KM) & (coverage.ymax >= HALF_KM))
print(f"Climate DT cell spacing: {CDT_CELL_KM:.1f} km -> about {int((WINDOW_KM / CDT_CELL_KM) ** 2)} cells per window")
print(f"window fully covered in {int(ok.sum())} of {len(ok)} display hours; cells inside: "
      f"min {coverage.n_in_window.min()}, median {int(coverage.n_in_window.median())}, max {coverage.n_in_window.max()}")
if area_files["sfc"]:
    with xr.open_dataset(next(iter(area_files["sfc"].values()))) as ds:
        print(f"area chunk: {ds.sizes['points']} cells, {ds.latitude.values.min():.2f}..{ds.latitude.values.max():.2f} N, "
              f"{ds.longitude.values.min():.2f}..{ds.longitude.values.max():.2f} E (map area {AREA_S}..{AREA_N} N, {AREA_W}..{AREA_E} E)")
```

### 5.2 Window cells (display period) and window statistics (all days)

Hourly `sfc` fields are matched to the ship position of the same hour; the daily `o2d` field
is valid for all 24 hours of its day and uses the 12 UTC position. Cells are kept only for
the display period (they feed the window viewers); for every day in the statistics range only
the window summary is stored.

```python
def cdt_values(ds, param):
    var, conv = PARAMS[param]["cdt"]
    if isinstance(var, tuple):                       # wind speed from components
        if var[0] not in ds:
            return None
        vals = np.hypot(ds[var[0]], ds[var[1]])
    else:
        if var not in ds:
            return None
        vals = ds[var]
    return conv(vals) if conv else vals


def cdt_day(path):
    # (arrays per param with shape (time, points), lat, lon, times) for one chunk file
    with xr.open_dataset(path) as ds:
        ds = ds.load()
    arrays = {}
    for p, spec in PARAMS.items():
        if spec["cdt"] is None:
            continue
        v = cdt_values(ds, p)
        if v is not None:
            arrays[p] = np.asarray(v.values, dtype=float).reshape(ds.sizes["time"], -1)
    return arrays, ds.latitude.values, ds.longitude.values, [pd.Timestamp(t) for t in ds.time.values]


def cdt_window(keep_cells_for):
    cells, summaries = [], []
    for group, files in window_files.items():
        daily = group == "o2d"
        for day, path in files.items():
            arrays, lat, lon, times = cdt_day(path)
            if not arrays:
                continue
            steps = [(0, day)] if daily else [(i, t) for i, t in enumerate(times) if t in stats_hours]
            for i, t in steps:
                if daily and not (stats_hours.normalize() == day).any():
                    continue
                x, y = to_window_xy(lat, lon, *ship_at(t + pd.Timedelta(hours=12) if daily else t))
                m = in_window(x, y)
                if not m.any():
                    continue
                for p, arr in arrays.items():
                    v = arr[i][m]
                    s = summarise(v)
                    if s:
                        summaries.append(dict(source="climate-dt", param=p, time=t, **s))
                    if t in keep_cells_for or (daily and (keep_cells_for.normalize() == day).any()):
                        cells.append(pd.DataFrame({"source": "climate-dt", "param": p, "time": t, "x_km": x[m], "y_km": y[m],
                                                   "lat": lat[m], "lon": lon[m], "cell_km": CDT_CELL_KM, "value": v}))
    cells = pd.concat(cells, ignore_index=True) if cells else pd.DataFrame(columns=WINDOW_COLUMNS)
    return cells, pd.DataFrame(summaries, columns=STAT_COLUMNS)


t0 = time.time()
win_cdt, stats_cdt = cdt_window(hours)
print(f"{len(win_cdt):,} Climate DT window cells (display period); {len(stats_cdt):,} window summaries over "
      f"{stats_cdt.time.dt.normalize().nunique()} days ({time.time() - t0:.0f} s)")
```

## 6. CARRA2: local pan-Arctic files cropped to the window

The Climate Data Store delivers CARRA2 on the full 2869 × 2869 Lambert grid (the `area`
keyword is not applied for this dataset), so each file is first cut to an index box around
the track positions of its own times and then cropped analysis time by analysis time. Every
netCDF under `~/Git/polarhack/data/carra2` and `data/spatial-window/carra2` is scanned and
variables are recognised by name, so files retrieved with the cell below join automatically.
Statistics use every analysis time on disk; cells are kept for the display period.

```python
if FETCH_CARRA2:
    # Retrieve more CARRA2 variables or months. Valid names come from the CDS form of
    # `reanalysis-pan-carra` (single_levels, analysis, 3-hourly). Expect a queue and ~250 MB
    # per variable-month; the area keyword is ignored by the CDS for this dataset.
    import cdsapi

    for (year, month), sub in sel.groupby([sel.index.year, sel.index.month]):
        target = CACHE / "carra2" / f"carra2_sfc_{year}{month:02d}.nc"
        if target.exists():
            print("exists:", target)
            continue
        request = {
            "level_type": "single_levels",
            "variable": ["2m_temperature", "10m_wind_speed", "mean_sea_level_pressure",
                         "total_column_cloud_liquid_water", "total_column_integrated_water_vapour",
                         "sea_ice_area_fraction"],
            "product_type": "analysis",
            "time": ["00:00", "03:00", "06:00", "09:00", "12:00", "15:00", "18:00", "21:00"],
            "year": [str(year)], "month": [f"{month:02d}"],
            "day": [f"{d:02d}" for d in sorted(sub.index.day.unique())],
            "data_format": "netcdf",
        }
        cdsapi.Client().retrieve("reanalysis-pan-carra", request).download(str(target))
        print("saved", target)
else:
    print("FETCH_CARRA2 is False: using the CARRA2 files already on disk")
```

```python
def carra_param_for(varname):
    for p, spec in PARAMS.items():
        if spec.get("carra") and varname in spec["carra"][0]:
            return p
    return None


def carra_files():
    return sorted(f for d in CARRA2_DIRS for f in d.glob("*.nc") if "ArcLeads" not in f.name)


def open_carra(f):
    ds = xr.open_dataset(f)
    params = {v: carra_param_for(v) for v in ds.data_vars}
    params = {v: p for v, p in params.items() if p}
    if not params or "latitude" not in ds.coords:
        return None
    tname = "valid_time" if "valid_time" in ds.coords and ds["valid_time"].ndim == 1 else "time"
    times = pd.DatetimeIndex(ds[tname].values)
    return ds, params, times


def carra_window(keep_cells_for):
    cells, summaries = [], []
    for f in carra_files():
        opened = open_carra(f)
        if opened is None:
            continue
        ds, params, times = opened
        keep = [i for i, t in enumerate(times) if t in stats_hours]
        if not keep:
            continue
        pos = track.loc[times[keep]]
        lat2d = ds.latitude.values
        lon2d = (ds.longitude.values + 180.0) % 360.0 - 180.0
        box = ((lat2d >= pos.lat.min() - 1.0) & (lat2d <= pos.lat.max() + 1.0)
               & (lon2d >= pos.lon.min() - 8.0) & (lon2d <= pos.lon.max() + 8.0))
        jy, jx = np.where(box)
        ys, xs = slice(jy.min(), jy.max() + 1), slice(jx.min(), jx.max() + 1)
        ydim, xdim = ds.latitude.dims
        lat_c, lon_c = lat2d[ys, xs], lon2d[ys, xs]
        cell = float(np.hypot((lat_c[0, 1] - lat_c[0, 0]) * 111.2,
                              (lon_c[0, 1] - lon_c[0, 0]) * 111.2 * np.cos(np.radians(lat_c[0, 0]))))
        for var, p in params.items():
            sub = ds[var].isel({ydim: ys, xdim: xs}).isel({ds[var].dims[0]: keep}).load()
            conv = PARAMS[p]["carra"][1]
            for k, i in enumerate(keep):
                t = times[i]
                x, y = to_window_xy(lat_c.ravel(), lon_c.ravel(), *ship_at(t))
                m = in_window(x, y)
                v = np.asarray(sub.isel({sub.dims[0]: k}).values, dtype=float).ravel()[m]
                v = conv(v) if conv else v
                s = summarise(v)
                if s:
                    summaries.append(dict(source="carra2", param=p, time=t, **s))
                if t in keep_cells_for:
                    cells.append(pd.DataFrame({"source": "carra2", "param": p, "time": t, "x_km": x[m], "y_km": y[m],
                                               "lat": lat_c.ravel()[m], "lon": lon_c.ravel()[m], "cell_km": cell, "value": v}))
        print(f"{f.name}: {sorted(set(params.values()))}, {len(keep)} analysis times, cell {cell:.2f} km")
    cells = pd.concat(cells, ignore_index=True) if cells else pd.DataFrame(columns=WINDOW_COLUMNS)
    return cells, pd.DataFrame(summaries, columns=STAT_COLUMNS)


win_carra, stats_carra = carra_window(hours)
print(f"{len(win_carra):,} CARRA2 window cells (display period); {len(stats_carra):,} window summaries")
```

## 7. MOSAiC spatial observations: ArcLeads lead maps

ArcLeads (Willmes et al.) classifies MODIS thermal-infrared imagery into daily 1 km maps
(0 cloud, 1 land, 2 sea ice, 3 artefact, 4 lead, 5 open water) for November to April. Per day
and window two layers are derived: `lead_map`, the raw classes on the 1 km grid, and
`lead_fraction`, (leads + open water) / (sea ice + leads + open water) on 5 km cells, left
empty where clouds or artefacts hide more than half of a cell. Both are centred on the ship
position at 12 UTC. Statistics use every ArcLeads day in the statistics range, cells are kept
for the display period.

```python
def arcleads_block(ds, lat0, lon0, half_km, stride=20):
    # index slices of the 1 km grid covering half_km around (lat0, lon0), found on a strided subsample
    lat_s = ds.lat[::stride, ::stride].values.astype(float)
    lon_s = ds.lon[::stride, ::stride].values.astype(float)
    xs_, ys_ = to_window_xy(lat_s.ravel(), lon_s.ravel(), lat0, lon0)
    cy, cx = np.unravel_index(int(np.argmin(np.hypot(xs_, ys_))), lat_s.shape)
    cy, cx, block = cy * stride, cx * stride, int(half_km * 1.6)
    ny, nx = ds.lat.shape
    return slice(max(cy - block, 0), min(cy + block, ny)), slice(max(cx - block, 0), min(cx + block, nx))


def lead_fraction_grid(x, y, classes, agg_km, half_km):
    gx = np.floor((x + half_km) / agg_km).astype(int)
    gy = np.floor((y + half_km) / agg_km).astype(int)
    frame = pd.DataFrame({"gx": gx, "gy": gy, "valid": np.isin(classes, [2, 4, 5]), "open": np.isin(classes, [4, 5])})
    g = frame.groupby(["gx", "gy"]).agg(n=("valid", "size"), valid=("valid", "sum"), open=("open", "sum"))
    frac = np.where(g.valid >= 0.5 * g.n, g.open / g.valid.clip(lower=1), np.nan)
    cx = (g.index.get_level_values("gx") + 0.5) * agg_km - half_km
    cy = (g.index.get_level_values("gy") + 0.5) * agg_km - half_km
    return np.asarray(cx), np.asarray(cy), frac


def arcleads_window(keep_cells_for, agg_km=5.0):
    cells, summaries = [], []
    for f in ARCLEADS_FILES:
        ds = xr.open_dataset(f)
        ftimes = pd.to_datetime(ds.time.values.astype(str), format="%Y%m%d")
        idx = [i for i, t in enumerate(ftimes) if t in stats_days]
        if not idx:
            continue
        for i in idx:
            day = ftimes[i]
            lat0, lon0 = ship_at(day + pd.Timedelta(hours=12))
            ys, xs = arcleads_block(ds, lat0, lon0, HALF_KM)
            lat_c = ds.lat.isel(nrows=ys, ncols=xs).values.astype(float)
            lon_c = ds.lon.isel(nrows=ys, ncols=xs).values.astype(float)
            classes = ds.leadmap.isel(time_dim=i, nrows=ys, ncols=xs).values.astype(float).ravel()
            x, y = to_window_xy(lat_c.ravel(), lon_c.ravel(), lat0, lon0)
            m = in_window(x, y)
            if not m.any():
                continue
            cell = float(np.hypot(x[1] - x[0], y[1] - y[0]))
            cx, cy, frac = lead_fraction_grid(x[m], y[m], classes[m], agg_km, HALF_KM)
            s = summarise(frac)
            if s:
                summaries.append(dict(source="mosaic", param="lead_fraction", time=day, **s))
            if day in keep_cells_for.normalize():
                cells.append(pd.DataFrame({"source": "mosaic", "param": "lead_map", "time": day, "x_km": x[m], "y_km": y[m],
                                           "lat": lat_c.ravel()[m], "lon": lon_c.ravel()[m], "cell_km": cell, "value": classes[m]}))
                cells.append(pd.DataFrame({"source": "mosaic", "param": "lead_fraction", "time": day, "x_km": cx, "y_km": cy,
                                           "lat": np.nan, "lon": np.nan, "cell_km": agg_km, "value": frac}))
        print(f"{f.name}: {len(idx)} days in the statistics range")
    cells = pd.concat(cells, ignore_index=True) if cells else pd.DataFrame(columns=WINDOW_COLUMNS)
    return cells, pd.DataFrame(summaries, columns=STAT_COLUMNS)


t0 = time.time()
win_leads, stats_leads = arcleads_window(hours)
print(f"{len(win_leads):,} ArcLeads window cells (display period); {len(stats_leads):,} daily summaries ({time.time() - t0:.0f} s)")
```

## 8. One table for the cells, one for the statistics

`win` holds every window cell of the display period; `stats` holds the window mean, spread,
extremes and cell count per dataset, parameter and time over the whole statistics range,
joined with the ship observation at that time. Both are cached as compressed CSV next to the
notebook.

```python
win = pd.concat([win_cdt, win_carra, win_leads], ignore_index=True)
win["time"] = pd.to_datetime(win["time"])
win = win.dropna(subset=["value"]).reset_index(drop=True)

stats = pd.concat([stats_cdt, stats_carra, stats_leads], ignore_index=True)
stats["time"] = pd.to_datetime(stats["time"])
obs_long = obs.stack().rename("obs").reset_index()
obs_long.columns = ["time", "param", "obs"]
stats = stats.merge(obs_long, on=["time", "param"], how="left").sort_values(["source", "param", "time"]).reset_index(drop=True)
stats["obs"] = np.where(stats.param.isin(DAILY_PARAMS), np.nan, stats["obs"])

win.to_csv(CACHE / "window_cells.csv.gz", index=False)
stats.to_csv(CACHE / "window_stats.csv.gz", index=False)
print(f"{len(win):,} window cells (display period), {len(stats):,} window summaries")

print("\ncells per dataset and parameter in the display period:")
display(win.groupby(["source", "param"]).agg(times=("time", "nunique"), cells_per_time=("time", lambda s: int(len(s) / s.nunique()))))
print("statistics coverage along the track:")
display(stats.groupby(["source", "param"]).agg(first=("time", "min"), last=("time", "max"), times=("time", "size")))
```

## 9. Window viewers (window coordinates)

Two dropdowns choose the **dataset** and **parameter**, the slider moves the window along
the track. For 3-hourly (CARRA2) and daily (ArcLeads, sea ice) layers the nearest available
field is shown and its valid time is written in the title. The right-hand panel shows the
window mean (line) and ±1 spatial standard deviation (band) of the selected dataset over the
**whole statistics range**, the other datasets' means, the ship observation, and a cursor at
the slider time.

```python
import contextlib

import ipywidgets as widgets
import plotly.graph_objects as go
from plotly.subplots import make_subplots

_nullcontext = contextlib.nullcontext
try:
    import anywidget  # noqa: F401  (needed by go.FigureWidget for in-place updates)
    HAVE_FIGUREWIDGET = True
except ImportError:
    HAVE_FIGUREWIDGET = False

HOURLY = list(hours)
PX_PER_KM = 4.2

CRANGE = {p: (float(g.value.quantile(0.02)), float(g.value.quantile(0.98))) for p, g in win.groupby("param")}
CRANGE["lead_map"] = (0, 5)
CRANGE["lead_fraction"] = (0, max(0.05, CRANGE.get("lead_fraction", (0, 0.2))[1]))
CRANGE["siconc"] = (min(0.8, CRANGE.get("siconc", (0.8, 1))[0]), 1.0)


def params_for(source, table=None):
    table = win if table is None else table
    return sorted(table.loc[table.source == source, "param"].unique(), key=PARAM_ORDER.index)


def nearest_time(times, t, param):
    times = pd.DatetimeIndex(sorted(times))
    if len(times) == 0:
        return None
    if param in DAILY_PARAMS:
        same_day = times[times.normalize() == pd.Timestamp(t).normalize()]
        return same_day[0] if len(same_day) else None
    i = times.get_indexer([pd.Timestamp(t)], method="nearest")[0]
    return times[i] if abs(times[i] - pd.Timestamp(t)) <= pd.Timedelta(hours=1.5) else None


def nearest_field_time(source, param, t):
    return nearest_time(win.loc[(win.source == source) & (win.param == param), "time"].unique(), t, param)


def field(source, param, t):
    return win[(win.source == source) & (win.param == param) & (win.time == t)]


def track_segment(t, hours_before=24, hours_after=24):
    lat0, lon0 = ship_at(t)
    seg = track.loc[pd.Timestamp(t) - pd.Timedelta(hours=hours_before): pd.Timestamp(t) + pd.Timedelta(hours=hours_after)]
    return to_window_xy(seg.lat.values, seg.lon.values, lat0, lon0)


def build_figure():
    fig = make_subplots(rows=1, cols=2, column_widths=[0.42, 0.58], horizontal_spacing=0.08,
                        subplot_titles=("window", "window statistics along the drift"))
    fig.add_trace(go.Scattergl(x=[], y=[], mode="markers", name="field", showlegend=False,
                               marker=dict(symbol="square", size=6, colorscale="Viridis", showscale=True,
                                           colorbar=dict(x=0.40, len=0.9, thickness=14, title=""))), row=1, col=1)
    fig.add_trace(go.Scatter(x=[], y=[], mode="lines", name="track ±24 h", line=dict(color="black", width=1.5), showlegend=False), row=1, col=1)
    fig.add_trace(go.Scatter(x=[0], y=[0], mode="markers", name="Polarstern", showlegend=False,
                             marker=dict(symbol="star", size=14, color="white", line=dict(color="black", width=1.5))), row=1, col=1)
    fig.add_trace(go.Scatter(x=[], y=[], mode="lines", line=dict(width=0), showlegend=False, hoverinfo="skip"), row=1, col=2)
    fig.add_trace(go.Scatter(x=[], y=[], mode="lines", line=dict(width=0), fill="tonexty",
                             fillcolor="rgba(120,120,120,0.25)", name="±1 σ (window)", hoverinfo="skip"), row=1, col=2)
    for s in SOURCES:
        fig.add_trace(go.Scatter(x=[], y=[], mode="lines", name=f"{s} window mean", line=dict(color=SOURCE_COLORS[s], width=1.2)), row=1, col=2)
    fig.add_trace(go.Scatter(x=[], y=[], mode="lines", name="MOSAiC obs at ship", line=dict(color="black", width=1.0, dash="dot")), row=1, col=2)
    fig.update_xaxes(title_text="east (km)", range=[-HALF_KM, HALF_KM], constrain="domain", row=1, col=1)
    fig.update_yaxes(title_text="north (km)", range=[-HALF_KM, HALF_KM], scaleanchor="x", scaleratio=1, row=1, col=1)
    fig.update_xaxes(title_text="time (UTC)", row=1, col=2)
    fig.update_layout(height=520, width=1050, margin=dict(l=50, r=20, t=70, b=50),
                      legend=dict(orientation="h", y=-0.18, x=0.4), template="plotly_white")
    fig.add_shape(type="line", x0=0, x1=0, y0=0, y1=1, xref="x2", yref="paper", line=dict(color="crimson", width=1.5, dash="dash"))
    return fig


def fill_figure(fig, source, param, t, fixed_range=True, show_track=True):
    t = pd.Timestamp(t)
    ft = nearest_field_time(source, param, t)
    spec = PARAMS[param]
    df = field(source, param, ft) if ft is not None else win.iloc[0:0]
    size = float(df.cell_km.iloc[0]) * PX_PER_KM * 0.98 if len(df) else 6
    lo, hi = CRANGE[param] if fixed_range else ((float(df.value.min()), float(df.value.max())) if len(df) else (0, 1))
    seg_x, seg_y = track_segment(t) if show_track else ([], [])
    st = stats[(stats.source == source) & (stats.param == param)].sort_values("time")
    with (fig.batch_update() if HAVE_FIGUREWIDGET else _nullcontext()):
        fig.data[0].update(x=df.x_km, y=df.y_km,
                           marker=dict(color=df.value, cmin=lo, cmax=hi, size=max(size, 2.5), colorscale=spec["cmap"],
                                       symbol="square", showscale=True,
                                       colorbar=dict(x=0.40, len=0.9, thickness=14, title=dict(text=spec["units"]))),
                           hovertemplate="x %{x:.1f} km, y %{y:.1f} km<br>%{marker.color:.3g} " + spec["units"] + "<extra></extra>")
        fig.data[1].update(x=seg_x, y=seg_y)
        fig.data[3].update(x=st.time, y=st["mean"] - st["std"])
        fig.data[4].update(x=st.time, y=st["mean"] + st["std"])
        for k, s in enumerate(SOURCES):
            ss = stats[(stats.source == s) & (stats.param == param)].sort_values("time")
            fig.data[5 + k].update(x=ss.time, y=ss["mean"], visible=True if len(ss) else "legendonly")
        o = obs[param] if param in obs else pd.Series(dtype=float)
        fig.data[8].update(x=o.index, y=o.values, visible=True if o.notna().any() else "legendonly")
        fig.layout.shapes[0].update(x0=t, x1=t)
        valid = f"{ft:%Y-%m-%d %H:%M}" if ft is not None else "no field within tolerance"
        stat_txt = (f"mean {df.value.mean():.3g}, σ {df.value.std():.3g}, min {df.value.min():.3g}, max {df.value.max():.3g} {spec['units']}"
                    if len(df) else "")
        obs_val = obs.at[t, param] if (param in obs and t in obs.index) else np.nan
        obs_txt = f" | ship obs {obs_val:.3g} {spec['units']}" if np.isfinite(obs_val) else ""
        lat0, lon0 = ship_at(t)
        fig.layout.title.text = (f"<b>{SOURCES[source]}</b> — {spec['label']}<br>"
                                 f"<sup>window {WINDOW_KM:.0f}×{WINDOW_KM:.0f} km at {lat0:.2f}°N {lon0:.2f}°E, "
                                 f"{t:%Y-%m-%d %H:%M} UTC | field valid {valid} | {len(df)} cells: {stat_txt}{obs_txt}</sup>")
        fig.layout.yaxis2.title.text = f"{spec['label']} ({spec['units']})"
    return fig


w_source = widgets.Dropdown(options=[(SOURCES[s], s) for s in SOURCES if s in win.source.unique()],
                            description="dataset", layout=widgets.Layout(width="360px"))
w_param = widgets.Dropdown(options=[(PARAMS[p]["label"], p) for p in params_for(w_source.value)],
                           description="parameter", layout=widgets.Layout(width="420px"))
w_time = widgets.SelectionSlider(options=[(f"{t:%Y-%m-%d %H:%M}", t) for t in HOURLY], description="time",
                                 continuous_update=False, layout=widgets.Layout(width="1000px"))
w_fixed = widgets.Checkbox(value=True, description="fixed colour range", indent=False)
w_track = widgets.Checkbox(value=True, description="show track ±24 h", indent=False)
w_play = widgets.Play(interval=600, min=0, max=len(HOURLY) - 1, step=3, description="play")
widgets.jslink((w_play, "value"), (w_time, "index"))

fig = go.FigureWidget(build_figure()) if HAVE_FIGUREWIDGET else build_figure()
out = widgets.Output()


def link_param_dropdown(source_dd, param_dd, table=None):
    def on_change(change):
        opts = [(PARAMS[p]["label"], p) for p in params_for(change["new"], table)]
        current = param_dd.value
        param_dd.options = opts
        param_dd.value = current if current in [p for _, p in opts] else opts[0][1]
    source_dd.observe(on_change, names="value")


def redraw(*_):
    fill_figure(fig, w_source.value, w_param.value, w_time.value, w_fixed.value, w_track.value)
    if not HAVE_FIGUREWIDGET:
        with out:
            out.clear_output(wait=True)
            fig.show()


link_param_dropdown(w_source, w_param)
for w in (w_source, w_param, w_time, w_fixed, w_track):
    w.observe(redraw, names="value")
start_t = next((t for t in HOURLY if t.hour == 12), HOURLY[0])
w_time.value = start_t
redraw()
display(widgets.VBox([widgets.HBox([w_source, w_param, w_fixed, w_track]), widgets.HBox([w_play, w_time]),
                      fig if HAVE_FIGUREWIDGET else out]))
```

### 9.1 All datasets side by side

Same parameter, same slider time, one window per dataset, to compare how the 12.7 km
Climate DT cells, the 2.5 km CARRA2 grid and the 1 km lead map resolve the same area.

```python
def side_by_side(param, t, fixed_range=True):
    t = pd.Timestamp(t)
    sources = [s for s in SOURCES if len(win[(win.source == s) & (win.param == param)])]
    fig = make_subplots(rows=1, cols=max(len(sources), 1), horizontal_spacing=0.06, subplot_titles=[SOURCES[s] for s in sources])
    lo, hi = CRANGE[param]
    for k, s in enumerate(sources, start=1):
        ft = nearest_field_time(s, param, t)
        df = field(s, param, ft) if ft is not None else win.iloc[0:0]
        if not fixed_range and len(df):
            lo, hi = float(df.value.min()), float(df.value.max())
        size = float(df.cell_km.iloc[0]) * 3.0 * 0.98 if len(df) else 6
        fig.add_trace(go.Scattergl(x=df.x_km, y=df.y_km, mode="markers", showlegend=False,
                                   marker=dict(symbol="square", size=max(size, 2.5), color=df.value, cmin=lo, cmax=hi,
                                               colorscale=PARAMS[param]["cmap"], showscale=(k == len(sources)),
                                               colorbar=dict(title=PARAMS[param]["units"], thickness=14)),
                                   hovertemplate="%{marker.color:.3g}<extra>" + s + "</extra>"), row=1, col=k)
        sx, sy = track_segment(t)
        fig.add_trace(go.Scatter(x=sx, y=sy, mode="lines", line=dict(color="black", width=1), showlegend=False), row=1, col=k)
        fig.add_trace(go.Scatter(x=[0], y=[0], mode="markers", showlegend=False,
                                 marker=dict(symbol="star", size=12, color="white", line=dict(color="black", width=1.5))), row=1, col=k)
        valid = f"valid {ft:%Y-%m-%d %H:%M}" if ft is not None else "no field"
        fig.layout.annotations[k - 1].text = f"{SOURCES[s]}<br><sup>{valid}, {len(df)} cells</sup>"
        fig.update_xaxes(range=[-HALF_KM, HALF_KM], title_text="east (km)", row=1, col=k)
        fig.update_yaxes(range=[-HALF_KM, HALF_KM], scaleanchor=f"x{k if k > 1 else ''}", title_text="north (km)" if k == 1 else None, row=1, col=k)
    lat0, lon0 = ship_at(t)
    fig.update_layout(height=380, width=330 * max(len(sources), 1) + 80, template="plotly_white",
                      title=f"{PARAMS[param]['label']} — {t:%Y-%m-%d %H:%M} UTC, window at {lat0:.2f}°N {lon0:.2f}°E", margin=dict(t=90, b=50))
    return fig


w2_param = widgets.Dropdown(options=[(PARAMS[p]["label"], p) for p in sorted(win.param.unique(), key=PARAM_ORDER.index)],
                            description="parameter", layout=widgets.Layout(width="420px"))
w2_time = widgets.SelectionSlider(options=[(f"{t:%Y-%m-%d %H:%M}", t) for t in HOURLY], description="time",
                                  value=start_t, continuous_update=False, layout=widgets.Layout(width="1000px"))
out2 = widgets.Output()


def redraw2(*_):
    with out2:
        out2.clear_output(wait=True)
        side_by_side(w2_param.value, w2_time.value).show()


w2_param.observe(redraw2, names="value")
w2_time.observe(redraw2, names="value")
redraw2()
display(widgets.VBox([widgets.HBox([w2_param]), w2_time, out2]))
```

## 10. Window statistics along the full track

`stats` holds the window mean per dataset and time next to the observation at the ship for
the whole statistics range. The first table compares each dataset with the observations on
the times where both exist; the second summarises the *spatial* variability inside the window
(mean σ), which tells how representative a single point is for the 100 km domain. The figure
shows the full-track time series of window mean ± σ for a chosen parameter together with the
ship observation; the shaded band marks the display period. Coverage differs by dataset:
Climate DT covers every drift day, CARRA2 the months on disk (October 2019 unless more is
retrieved), ArcLeads November to April, the observations 4 Oct 2019 – 19 Sep 2020.

```python
def score(g):
    d = g.dropna(subset=["mean", "obs"])
    if len(d) < 3:
        return pd.Series(dict(n=len(d), bias=np.nan, rmsd=np.nan, r=np.nan))
    diff = d["mean"] - d["obs"]
    return pd.Series(dict(n=len(d), bias=diff.mean(), rmsd=np.sqrt((diff ** 2).mean()), r=d["mean"].corr(d["obs"])))


scores = (stats[~stats.param.isin(DAILY_PARAMS)].groupby(["param", "source"]).apply(score, include_groups=False)
          .dropna(subset=["bias"]).round(3))
scores.columns = ["n (times)", "bias (window mean − obs)", "RMSD", "r"]
display(scores)

spatial = stats.assign(rng=stats["max"] - stats["min"]).groupby(["param", "source"]).agg(
    mean_sigma=("std", "mean"), max_sigma=("std", "max"), mean_range=("rng", "mean"),
    first=("time", "min"), last=("time", "max"), n_times=("time", "nunique"))
spatial.columns = ["mean σ in window", "max σ in window", "mean (max − min)", "first", "last", "n times"]
display(spatial.round(3))
```

```python
def rgba(hex_color, alpha):
    r, g, b = (int(hex_color[i:i + 2], 16) for i in (1, 3, 5))
    return f"rgba({r},{g},{b},{alpha})"


def stats_timeseries(param, resample=None):
    # full-track window mean ± spatial σ per dataset and the ship observation; resample e.g. "1D" to smooth
    fig = go.Figure()
    for s in SOURCES:
        ss = stats[(stats.source == s) & (stats.param == param)].set_index("time").sort_index()
        if not len(ss):
            continue
        if resample:
            ss = ss.resample(resample).mean(numeric_only=True).dropna(subset=["mean"])
        fig.add_trace(go.Scatter(x=pd.concat([pd.Series(ss.index), pd.Series(ss.index[::-1])]),
                                 y=pd.concat([ss["mean"] + ss["std"], (ss["mean"] - ss["std"])[::-1]]),
                                 fill="toself", fillcolor=rgba(SOURCE_COLORS[s], 0.18), line=dict(width=0), showlegend=False, hoverinfo="skip"))
        fig.add_trace(go.Scatter(x=ss.index, y=ss["mean"], mode="lines", name=f"{s} window mean ± σ ({len(ss)} times)",
                                 line=dict(color=SOURCE_COLORS[s], width=1.2)))
    if param in obs:
        o = obs[param].resample(resample).mean() if resample else obs[param]
        fig.add_trace(go.Scatter(x=o.index, y=o.values, mode="lines", name="MOSAiC obs at ship", line=dict(color="black", width=1.0, dash="dot")))
    fig.add_vrect(x0=hours[0], x1=hours[-1], fillcolor="grey", opacity=0.12, line_width=0, annotation_text="display period", annotation_position="top left")
    fig.update_layout(template="plotly_white", height=420, width=1050,
                      title=f"{PARAMS[param]['label']}: {WINDOW_KM:.0f} km window mean ± spatial σ along the full drift"
                            + (f" ({resample} means)" if resample else ""),
                      yaxis_title=f"{PARAMS[param]['label']} ({PARAMS[param]['units']})", legend=dict(orientation="h", y=-0.2),
                      xaxis=dict(range=[track.index[0], track.index[-1]], title="time (UTC)"))
    return fig


stats_timeseries("t2m").show()
stats_timeseries("wspd10", resample="1D").show()
```

## 11. Fixed polar map: the window moving along the track

A fixed map of the area around the display-period track (`MAP_AREA`, north-polar
stereographic projection with the latitude/longitude graticule) shows the complete field of
the selected dataset and variable, with the 100 km window and the ship moving along the
track. Fields are drawn as filled cells, so there are no gaps between neighbours:

- Climate DT: the actual HEALPix nside-512 cell polygons (`healpy` boundaries) filled with
  the bounding-box download from section 5, hourly;
- CARRA2: the 2.5 km grid averaged to `MAP_GRID_KM` quadrilaterals, 3-hourly;
- MOSAiC: the ArcLeads lead + open-water fraction on the same `MAP_GRID_KM` grid, daily.

Dropdowns select the **dataset** and the **variable**; the play button and slider move the
window along the track. 11.3 is a kernel-free version of the same map with the choices in a
single menu.

### 11.1 Area fields on fixed cell geometries

```python
import healpy as hp

NSIDE = 512


def area_mask(lat, lon):
    return (lat >= AREA_S) & (lat <= AREA_N) & (lon >= AREA_W) & (lon <= AREA_E)


def load_cdt_area():
    # (pix -> lat/lon) master list plus per-param tables (time, pix) from the area chunk files
    tables = {}
    for group, files in area_files.items():
        for day, path in files.items():
            arrays, lat, lon, times = cdt_day(path)
            pix = hp.ang2pix(NSIDE, lon, lat, nest=True, lonlat=True)
            if group == "o2d":
                times = [day]
            for p, arr in arrays.items():
                for i, t in enumerate(times):
                    tables.setdefault(p, {})[t] = pd.Series(arr[i], index=pix)
    out = {}
    for p, rows in tables.items():
        out[p] = pd.DataFrame(rows).T.sort_index()          # index time, columns pix
    all_pix = sorted(set().union(*[set(df.columns) for df in out.values()])) if out else []
    return {p: df.reindex(columns=all_pix) for p, df in out.items()}, np.array(all_pix, dtype=np.int64)


def healpix_geojson(pix):
    b = hp.boundaries(NSIDE, pix, step=1, nest=True)      # (npix, 3, 4)
    feats = []
    for k, pid in enumerate(pix):
        blon, blat = hp.vec2ang(b[k].T, lonlat=True)
        blon = (blon + 180.0) % 360.0 - 180.0
        ring = [[float(a), float(c)] for a, c in zip(blon, blat)] + [[float(blon[0]), float(blat[0])]]
        feats.append({"type": "Feature", "id": int(pid), "geometry": {"type": "Polygon", "coordinates": [ring]}})
    return {"type": "FeatureCollection", "features": feats}


def quad_geojson(lat2d, lon2d, ids):
    # quadrilaterals around the cell centres of a 2-D grid: corners are midpoints between
    # neighbouring centres in the tangent plane of the map area (edges extrapolated)
    x, y = to_window_xy(lat2d, lon2d, AREA_LAT0, AREA_LON0)

    def corners(a):
        c = np.empty((a.shape[0] + 1, a.shape[1] + 1))
        c[1:-1, 1:-1] = 0.25 * (a[:-1, :-1] + a[1:, :-1] + a[:-1, 1:] + a[1:, 1:])
        c[0, 1:-1] = 2 * c[1, 1:-1] - c[2, 1:-1]
        c[-1, 1:-1] = 2 * c[-2, 1:-1] - c[-3, 1:-1]
        c[:, 0] = 2 * c[:, 1] - c[:, 2]
        c[:, -1] = 2 * c[:, -2] - c[:, -3]
        return c

    cx, cy = corners(x), corners(y)
    clat, clon = from_window_xy(cx, cy, AREA_LAT0, AREA_LON0)
    feats = []
    ny, nx = lat2d.shape
    for j in range(ny):
        for i in range(nx):
            ring = [[float(clon[j, i]), float(clat[j, i])], [float(clon[j, i + 1]), float(clat[j, i + 1])],
                    [float(clon[j + 1, i + 1]), float(clat[j + 1, i + 1])], [float(clon[j + 1, i]), float(clat[j + 1, i])]]
            feats.append({"type": "Feature", "id": int(ids[j, i]), "geometry": {"type": "Polygon", "coordinates": [ring + [ring[0]]]}})
    return {"type": "FeatureCollection", "features": feats}


def block_mean(a, k):
    ny, nx = (a.shape[-2] // k) * k, (a.shape[-1] // k) * k
    a = a[..., :ny, :nx]
    return np.nanmean(a.reshape(*a.shape[:-2], ny // k, k, nx // k, k), axis=(-3, -1))


def load_carra_area():
    tables, geo = {}, None
    k = max(int(round(MAP_GRID_KM / 2.5)), 1)
    for f in carra_files():
        opened = open_carra(f)
        if opened is None:
            continue
        ds, params, times = opened
        keep = [i for i, t in enumerate(times) if t in hours]
        if not keep:
            continue
        lat2d = ds.latitude.values
        lon2d = (ds.longitude.values + 180.0) % 360.0 - 180.0
        jy, jx = np.where(area_mask(lat2d, lon2d))
        ys, xs = slice(jy.min(), jy.max() + 1), slice(jx.min(), jx.max() + 1)
        ydim, xdim = ds.latitude.dims
        lat_c, lon_c = block_mean(lat2d[ys, xs], k), block_mean(lon2d[ys, xs], k)
        if geo is None:
            ids = np.arange(lat_c.size).reshape(lat_c.shape)
            geo = (quad_geojson(lat_c, lon_c, ids), ids.ravel(), lat_c, lon_c)
        for var, p in params.items():
            sub = ds[var].isel({ydim: ys, xdim: xs}).isel({ds[var].dims[0]: keep}).load().values.astype(float)
            conv = PARAMS[p]["carra"][1]
            sub = block_mean(conv(sub) if conv else sub, k).reshape(len(keep), -1)
            tables.setdefault(p, {}).update({times[i]: sub[n] for n, i in enumerate(keep)})
    out = {p: pd.DataFrame(rows).T.sort_index() for p, rows in tables.items()}
    return out, geo


def load_leads_area():
    # ArcLeads lead fraction on a regular MAP_GRID_KM grid in the map-area tangent plane
    tables, geo = {}, None
    ring_lat, ring_lon = np.array([AREA_N, AREA_N, AREA_S, AREA_S]), np.array([AREA_W, AREA_E, AREA_W, AREA_E])
    edge_lat = np.r_[np.full(50, AREA_N), np.full(50, AREA_S), np.linspace(AREA_S, AREA_N, 50), np.linspace(AREA_S, AREA_N, 50)]
    edge_lon = np.r_[np.linspace(AREA_W, AREA_E, 50), np.linspace(AREA_W, AREA_E, 50), np.full(50, AREA_W), np.full(50, AREA_E)]
    ex, ey = to_window_xy(edge_lat, edge_lon, AREA_LAT0, AREA_LON0)
    gx = np.arange(np.floor(ex.min() / MAP_GRID_KM), np.ceil(ex.max() / MAP_GRID_KM) + 1) * MAP_GRID_KM
    gy = np.arange(np.floor(ey.min() / MAP_GRID_KM), np.ceil(ey.max() / MAP_GRID_KM) + 1) * MAP_GRID_KM
    cxx, cyy = np.meshgrid(gx[:-1] + MAP_GRID_KM / 2, gy[:-1] + MAP_GRID_KM / 2)
    clat, clon = from_window_xy(cxx, cyy, AREA_LAT0, AREA_LON0)
    ids = np.arange(cxx.size).reshape(cxx.shape)
    half_x = max(abs(ex.min()), abs(ex.max()))
    half_y = max(abs(ey.min()), abs(ey.max()))
    for f in ARCLEADS_FILES:
        ds = xr.open_dataset(f)
        ftimes = pd.to_datetime(ds.time.values.astype(str), format="%Y%m%d")
        idx = [i for i, t in enumerate(ftimes) if t in days]
        if not idx:
            continue
        ys, xs = arcleads_block(ds, AREA_LAT0, AREA_LON0, max(half_x, half_y))
        lat_b = ds.lat.isel(nrows=ys, ncols=xs).values.astype(float).ravel()
        lon_b = ds.lon.isel(nrows=ys, ncols=xs).values.astype(float).ravel()
        x, y = to_window_xy(lat_b, lon_b, AREA_LAT0, AREA_LON0)
        ix = np.floor((x - gx[0]) / MAP_GRID_KM).astype(int)
        iy = np.floor((y - gy[0]) / MAP_GRID_KM).astype(int)
        inside = (ix >= 0) & (ix < cxx.shape[1]) & (iy >= 0) & (iy < cxx.shape[0])
        flat = (iy * cxx.shape[1] + ix)[inside]
        for i in idx:
            classes = ds.leadmap.isel(time_dim=i, nrows=ys, ncols=xs).values.astype(float).ravel()[inside]
            n = np.bincount(flat, minlength=cxx.size)
            valid = np.bincount(flat, weights=np.isin(classes, [2, 4, 5]), minlength=cxx.size)
            opn = np.bincount(flat, weights=np.isin(classes, [4, 5]), minlength=cxx.size)
            frac = np.where((n > 0) & (valid >= 0.5 * n), opn / np.maximum(valid, 1), np.nan)
            tables.setdefault("lead_fraction", {})[ftimes[i]] = frac
    geo = (quad_geojson(clat, clon, ids), ids.ravel(), clat, clon)
    out = {p: pd.DataFrame(rows).T.sort_index() for p, rows in tables.items()}
    return out, geo


t0 = time.time()
cdt_area, cdt_pix = load_cdt_area()
cdt_geo = healpix_geojson(cdt_pix) if len(cdt_pix) else None
carra_area, carra_geo = load_carra_area()
leads_area, leads_geo = load_leads_area()
AREA = {"climate-dt": (cdt_area, (cdt_geo, cdt_pix)), "carra2": (carra_area, carra_geo), "mosaic": (leads_area, leads_geo)}
for s, (tables, geo) in AREA.items():
    print(f"{s}: {len(geo[1]) if geo else 0} cells; " + ", ".join(f"{p} x{len(df)} times" for p, df in tables.items()))
print(f"built in {time.time() - t0:.0f} s")
```

### 11.2 Live viewer with dataset and variable dropdowns

```python
MAP_H, MAP_W = 680, 1000


def area_outline():
    lat = np.r_[np.full(40, AREA_N), np.linspace(AREA_N, AREA_S, 40), np.full(40, AREA_S), np.linspace(AREA_S, AREA_N, 40)]
    lon = np.r_[np.linspace(AREA_W, AREA_E, 40), np.full(40, AREA_E), np.linspace(AREA_E, AREA_W, 40), np.full(40, AREA_W)]
    return np.r_[lat, lat[0]], np.r_[lon, lon[0]]


def area_params(source):
    return sorted(AREA[source][0].keys(), key=PARAM_ORDER.index)


def area_field(source, param, t):
    tables, geo = AREA[source]
    if param not in tables:
        return None, None
    ft = nearest_time(tables[param].index, t, param)
    if ft is None:
        return None, None
    return tables[param].loc[ft].values, ft


def build_polar_map():
    out_lat, out_lon = area_outline()
    fig = go.Figure()
    fig.add_trace(go.Choropleth(geojson={"type": "FeatureCollection", "features": []}, locations=[], z=[], featureidkey="id",
                                marker_line_width=0, colorscale="Viridis", name="field",
                                colorbar=dict(x=1.0, len=0.75, thickness=14)))
    fig.add_trace(go.Scattergeo(lat=out_lat, lon=out_lon, mode="lines", name="map area", line=dict(color="#888", width=1, dash="dot"), hoverinfo="skip"))
    fig.add_trace(go.Scattergeo(lat=sel.lat.values, lon=sel.lon.values, mode="lines", name="track (display period)", line=dict(color="black", width=1.2), hoverinfo="skip"))
    fig.add_trace(go.Scattergeo(lat=[], lon=[], mode="lines", name="track ±24 h", line=dict(color="black", width=3), hoverinfo="skip"))
    fig.add_trace(go.Scattergeo(lat=[], lon=[], mode="lines", name="100 km window", line=dict(color="crimson", width=3), hoverinfo="skip"))
    fig.add_trace(go.Scattergeo(lat=[], lon=[], mode="markers", name="Polarstern",
                                marker=dict(symbol="star", size=15, color="white", line=dict(color="black", width=1.5))))
    fig.update_layout(height=MAP_H, width=MAP_W, margin=dict(l=0, r=0, t=70, b=0), template="plotly_white",
                      legend=dict(orientation="h", y=0.0, x=0.0, bgcolor="rgba(255,255,255,0.7)"),
                      geo=dict(projection=dict(type="stereographic", rotation=dict(lon=AREA_LON0, lat=90)), fitbounds="locations",
                               showland=True, landcolor="#e9e4d6", showocean=True, oceancolor="#dfeaf5", showcoastlines=True,
                               coastlinecolor="#6f6f6f", resolution=50, showframe=False,
                               lataxis=dict(showgrid=True, gridcolor="#b8b8b8", gridwidth=0.6, dtick=0.5),
                               lonaxis=dict(showgrid=True, gridcolor="#b8b8b8", gridwidth=0.6, dtick=5)))
    return fig


def fill_polar_map(fig, source, param, t, fixed_range=True, current_geo=[None]):
    t = pd.Timestamp(t)
    z, ft = area_field(source, param, t)
    tables, (geojson, ids) = AREA[source]
    spec = PARAMS[param]
    lat0, lon0 = ship_at(t)
    ring_lat, ring_lon = window_ring(t)
    seg = track.loc[t - pd.Timedelta(hours=24): t + pd.Timedelta(hours=24)]
    if z is None:
        z = np.full(len(ids), np.nan)
    finite = np.isfinite(z)
    lo, hi = CRANGE.get(param, (np.nanmin(z), np.nanmax(z))) if fixed_range else ((float(np.nanmin(z)), float(np.nanmax(z))) if finite.any() else (0, 1))
    with (fig.batch_update() if isinstance(fig, go.FigureWidget) else _nullcontext()):
        if current_geo[0] != source:
            fig.data[0].update(geojson=geojson)
            current_geo[0] = source
        fig.data[0].update(locations=ids[finite], z=z[finite], zmin=lo, zmax=hi, colorscale=spec["cmap"],
                           colorbar=dict(x=1.0, len=0.75, thickness=14, title=dict(text=spec["units"])),
                           hovertemplate="%{z:.3g} " + spec["units"] + "<extra></extra>")
        fig.data[3].update(lat=seg.lat.values, lon=seg.lon.values)
        fig.data[4].update(lat=ring_lat, lon=ring_lon)
        fig.data[5].update(lat=[lat0], lon=[lon0])
        valid = f"field valid {ft:%Y-%m-%d %H:%M}" if ft is not None else "no field within tolerance"
        st = stats[(stats.source == source) & (stats.param == param) & (stats.time == (ft if ft is not None else t))]
        win_txt = f" | window mean {st['mean'].iloc[0]:.3g} {spec['units']}" if len(st) else ""
        obs_val = obs.at[t, param] if (param in obs and t in obs.index) else np.nan
        obs_txt = f" | ship obs {obs_val:.3g} {spec['units']}" if np.isfinite(obs_val) else ""
        fig.layout.title.text = (f"<b>{SOURCES[source]}</b> — {spec['label']} — {t:%Y-%m-%d %H:%M} UTC<br>"
                                 f"<sup>window {WINDOW_KM:.0f}×{WINDOW_KM:.0f} km at {lat0:.2f}°N {lon0:.2f}°E | {valid}{win_txt}{obs_txt}</sup>")
    return fig


p_source = widgets.Dropdown(options=[(SOURCES[s], s) for s in SOURCES if AREA[s][0]], description="dataset", layout=widgets.Layout(width="360px"))
p_param = widgets.Dropdown(options=[(PARAMS[p]["label"], p) for p in area_params(p_source.value)], description="variable", layout=widgets.Layout(width="420px"))
p_time = widgets.SelectionSlider(options=[(f"{t:%Y-%m-%d %H:%M}", t) for t in HOURLY], value=start_t, description="time",
                                 continuous_update=False, layout=widgets.Layout(width="1000px"))
p_play = widgets.Play(interval=500, min=0, max=len(HOURLY) - 1, step=1, value=HOURLY.index(start_t))
widgets.jslink((p_play, "value"), (p_time, "index"))
p_fixed = widgets.Checkbox(value=True, description="fixed colour range", indent=False)

polar_fig = build_polar_map()
polar_fig = go.FigureWidget(polar_fig) if HAVE_FIGUREWIDGET else polar_fig
polar_out = widgets.Output()


def on_p_source(change):
    opts = [(PARAMS[p]["label"], p) for p in area_params(change["new"])]
    cur = p_param.value
    p_param.options = opts
    p_param.value = cur if cur in [p for _, p in opts] else opts[0][1]


def redraw_polar(*_):
    fill_polar_map(polar_fig, p_source.value, p_param.value, p_time.value, p_fixed.value)
    if not HAVE_FIGUREWIDGET:
        with polar_out:
            polar_out.clear_output(wait=True)
            polar_fig.show()


p_source.observe(on_p_source, names="value")
for w in (p_source, p_param, p_time, p_fixed):
    w.observe(redraw_polar, names="value")
redraw_polar()
display(widgets.VBox([widgets.HBox([p_source, p_param, p_fixed]), widgets.HBox([p_play, p_time]), polar_fig if HAVE_FIGUREWIDGET else polar_out]))
```

### 11.3 Kernel-free animation of the same map

One self-contained figure: a menu with every dataset – variable combination, play/pause and a
slider over 3-hourly steps of the display period (`ANIM_STEP_HOURS`). Every combination is a
choropleth trace on its fixed geometry, frames only carry the values, so the figure stays a
few MB and survives an HTML export.

```python
ANIM_STEP_HOURS = 3
anim_times = [t for t in HOURLY if t.hour % ANIM_STEP_HOURS == 0]
combos = [(s, p) for s in SOURCES if AREA[s][0] for p in area_params(s)]

fig_anim = go.Figure()
for k, (s, p) in enumerate(combos):
    geojson, ids = AREA[s][1]
    fig_anim.add_trace(go.Choropleth(geojson=geojson, locations=ids, z=np.full(len(ids), np.nan), featureidkey="id",
                                     marker_line_width=0, colorscale=PARAMS[p]["cmap"], zmin=CRANGE[p][0], zmax=CRANGE[p][1],
                                     visible=(k == 0), name=f"{s} {p}", colorbar=dict(x=1.0, len=0.75, thickness=14, title=PARAMS[p]["units"]),
                                     hovertemplate="%{z:.3g} " + PARAMS[p]["units"] + "<extra></extra>"))
n_c = len(combos)
out_lat, out_lon = area_outline()
fig_anim.add_trace(go.Scattergeo(lat=out_lat, lon=out_lon, mode="lines", line=dict(color="#888", width=1, dash="dot"), hoverinfo="skip"))
fig_anim.add_trace(go.Scattergeo(lat=sel.lat.values, lon=sel.lon.values, mode="lines", line=dict(color="black", width=1.2), hoverinfo="skip"))
fig_anim.add_trace(go.Scattergeo(lat=[], lon=[], mode="lines", line=dict(color="black", width=3), hoverinfo="skip"))
fig_anim.add_trace(go.Scattergeo(lat=[], lon=[], mode="lines", line=dict(color="crimson", width=3), hoverinfo="skip"))
fig_anim.add_trace(go.Scattergeo(lat=[], lon=[], mode="markers", marker=dict(symbol="star", size=15, color="white", line=dict(color="black", width=1.5))))


def anim_frame(t):
    lat0, lon0 = ship_at(t)
    ring_lat, ring_lon = window_ring(t)
    seg = track.loc[t - pd.Timedelta(hours=24): t + pd.Timedelta(hours=24)]
    data = []
    for s, p in combos:
        z, _ = area_field(s, p, t)
        z = np.full(len(AREA[s][1][1]), np.nan) if z is None else np.round(z, 3)
        data.append(go.Choropleth(z=z))
    data += [go.Scattergeo(lat=seg.lat.values, lon=seg.lon.values), go.Scattergeo(lat=ring_lat, lon=ring_lon), go.Scattergeo(lat=[lat0], lon=[lon0])]
    return data


fig_anim.frames = [go.Frame(data=anim_frame(t), traces=list(range(n_c)) + [n_c + 2, n_c + 3, n_c + 4], name=f"{t:%Y-%m-%d %H:%M}") for t in anim_times]
first = fig_anim.frames[0].data
for k in range(n_c):
    fig_anim.data[k].z = first[k].z
for j, k in enumerate((n_c + 2, n_c + 3, n_c + 4)):
    fig_anim.data[k].update(lat=first[n_c + j].lat, lon=first[n_c + j].lon)

slider_steps = [dict(method="animate", label=f.name[5:], args=[[f.name], dict(mode="immediate", frame=dict(duration=0, redraw=True), transition=dict(duration=0))])
                for f in fig_anim.frames]
menu = [dict(label=f"{SOURCES[s]} – {PARAMS[p]['label']}", method="restyle",
             args=[{"visible": [j == k for j in range(n_c)] + [True] * 5}]) for k, (s, p) in enumerate(combos)]
fig_anim.update_layout(height=MAP_H, width=MAP_W, margin=dict(l=0, r=0, t=60, b=0), template="plotly_white", showlegend=False,
                       title_text=f"{WINDOW_KM:.0f} km window along the track, {anim_times[0]:%Y-%m-%d} … {anim_times[-1]:%Y-%m-%d}",
                       geo=polar_fig.layout.geo.to_plotly_json(),
                       updatemenus=[
                           dict(type="buttons", showactive=False, x=0.01, y=0.06, xanchor="left", yanchor="bottom", direction="right",
                                buttons=[dict(label="▶ play", method="animate",
                                              args=[None, dict(frame=dict(duration=350, redraw=True), fromcurrent=True, transition=dict(duration=0))]),
                                         dict(label="❚❚ pause", method="animate", args=[[None], dict(mode="immediate", frame=dict(duration=0, redraw=False))])]),
                           dict(buttons=menu, direction="down", x=0.01, y=0.99, xanchor="left", yanchor="top", showactive=True)],
                       sliders=[dict(steps=slider_steps, active=0, x=0.12, len=0.86, y=0.0, pad=dict(t=0, b=0),
                                     currentvalue=dict(prefix="time: ", visible=True, xanchor="left"), font=dict(size=10))])
print(f"{len(fig_anim.frames)} frames, {n_c} dataset–variable combinations; figure ≈ {len(fig_anim.to_json()) / 1e6:.1f} MB")
fig_anim.show()
```

## Notes and caveats

- **Story-nudging** keeps the Climate DT large-scale circulation close to the reanalysis of the
  actual day, so a day-by-day comparison with observations is meaningful. Change
  `CDT_EXPERIMENT` to `cont` or `Tplus2.0K` for the storyline counterfactuals (same requests).
- **Authentication**: the DESP offline token in `~/.polytopeapirc` is written by
  `desp-authentication.py`; section 1.1 checks it with a Polytope collection listing and only
  asks for credentials when that fails. Note that the script also reads the `USER`
  environment variable, which on a Mac is the local login name, so pass the DESP username
  explicitly (`DESP_USER`) rather than relying on the prompt-less path.
- **Polytope requests**: the trajectory `inflation` with `inflate: box` spans from the point
  to point + inflation (the point is the south-west corner); the window requests shift the
  positions accordingly and 5.1 verifies the coverage. The area requests use the
  `boundingbox` feature with `[[north, west], [south, east]]`.
- **Why the map is gap-free now**: Climate DT is a HEALPix point cloud; drawing it with
  screen-aligned square markers left gaps and overlaps between neighbours. Section 11 draws
  the true HEALPix cell polygons from `healpy`, filled with the complete bounding-box download,
  so the field is continuous. The window viewers in section 9 still use markers in window
  coordinates.
- CARRA2 `msl` versus the MOSAiC `atmos_pressure_2m` are not the same quantity: the ship
  value is station pressure at 2 m and is not reduced to sea level.
- ArcLeads is a satellite product, not a MOSAiC in-situ measurement, and the classes are
  sensitive to clouds (class 0) and thin-ice detection. PANGAEA also publishes lead fractions
  within a 50 km radius around Polarstern from seven sensors (von Albedyll et al. 2024,
  <https://doi.org/10.1594/PANGAEA.963736>); the download is protected by a browser challenge,
  so it cannot be scripted here.
- **Coverage along the track**: Climate DT is fetched for every drift day; CARRA2 covers
  October 2019 unless more months are retrieved (`FETCH_CARRA2`); ArcLeads exists November to
  April; the ship observations run 4 Oct 2019 – 19 Sep 2020 with gaps at leg transitions.
- OpenStreetMap tiles stop at 85.05 °N (Web Mercator), hence the plotly polar-stereographic
  map. The map area follows `PERIOD`; set `MAP_AREA = (north, west, south, east)` to fix it.
- Cached artefacts under `data/spatial-window/`: one netCDF per Climate DT day and group for
  the window requests and one per day for the area requests, `window_cells.csv.gz` and
  `window_stats.csv.gz`. Delete them to force a rebuild.
